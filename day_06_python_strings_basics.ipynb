# Q4. Take a string from the user. Count the total number of vowels present in the string.

```python
sen = input('Enter string: ')
vowel = 0
for ch in sen:
  if ch in 'AEIOUaeiou':
    vowel +=1
print("Vowels:",vowel)
```

# Q5. Take a string from the user.
 Count:
 - uppercase letters
 - lowercase letters
 - digits

```python
sen = input('Enter string: ')
sen = sen.replace(' ','')
upper_c = 0
lower_c = 0
digit_c = 0
for ch in sen:
  if ch.isupper():
    upper_c += 1

  elif ch.isdigit():
    digit_c += 1

  elif ch.islower():
    lower_c += 1
print("Uppercase letters:",upper_c)
print("Lowercase letters:",lower_c)
print("Digits:",digit_c)
```

# Q7. Take a string from the user. Check whether the string is a palindrome.

 Example:
- madam → Palindrome
- level → Palindrome
-  hello → Not Palindrome

```python
word = input('Enter string: ')
word = word.lower()
rev = word[::-1]
if rev == word:
  print('Palindrome')
else:
  print('Not palindrom')
```

# Q8. Take a string from the user. Print characters present at even indexes only.


```python
word = input('Enter string: ')
for index in range(len(word)):
  if index%2 ==0:
    print(word[index])
```

# Q9. Take a sentence from the user. Count how many spaces are present in the sentence.

```python
sen = input('Enter string: ')
count = 0
for ch in sen:
  if ch == ' ':
    count += 1

print('Present spaces:',count)
```

# Q10. Take a string from the user. Find and print:
 - first character
 - middle character
 - last character

 Handle both odd and even length strings appropriately.

```python
word = input('Enter string: ')
dist = len(word)//2
middle = ''
if len(word)%2 == 1:
  middle = word[dist]
elif len(word)%2 == 0:
  middle = word[dist-1]+word[dist]

print('Frist character:',word[0])
print('Middle Character:',middle)
print('Last character:',word[-1])
```
