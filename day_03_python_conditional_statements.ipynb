# Q1. Take a number from the user. Check whether the number is:
 - positive
 - negative
 - zero



```python
num = int(input('Enter any number: '))
if num > 0:
  print("Positive")
elif num == 0:
  print("Zero")
else:
  print('Negative')
```

# Q4. Take marks from the user.
 Print grade using the following conditions:

 - Marks >= 90  -> Grade A
 - Marks >= 75  -> Grade B
 - Marks >= 60  -> Grade C
 - Marks < 60   -> Grade D

```python
marks = int(input('Enter your marks: '))
if marks>=90:
  print('Grade A')
elif marks >= 75:
  print('Grade B')
elif marks >= 60:
  print('Grade C')
elif marks < 60:
  print('Grade D')
else :
  print('Invalid input')
```

# Q5. Take three numbers from the user. Find the largest number using conditional statements.




```python
num1 = int(input('Enter first number: '))
num2 = int(input('Enter second number: '))
num3 = int(input('Enter third number: '))

largest = num1
if largest < num2:
  largest = num2
if largest < num3:
  largest = num3

print("Largest number :",largest)
```

# Q6. Take a year from the user. Check whether the year is a leap year or not.


```python
year = int(input("Enter year: "))
if (year%4==0 and year%100 !=0) or year % 400 == 0:
  print("Year is leap year")
else:
  print('Year is not leap year')
```

# Q8. Take a number from the user. Check whether it is divisible by:
 - both 3 and 5
- only 3
- only 5
- neither



```python
num = int(input('Enter any number: '))
if num%3 == 0 and num%5 == 0:
  print("Number is divisible by both 3 and 5")
elif num%3 == 0:
  print("Number is divisible by only 3")
elif num%5 == 0:
  print("Number is divisible by only 5")
else:
  print("Number is divisible by neither 3 and 5")
```


# Q9. Take electricity units consumed from the user. Calculate electricity bill using conditions.

 Rules:
- First 100 units -> 5 per unit
- Next 100 units -> 7 per unit
- Above 200 units -> 10 per unit




```python
units = int(input('Enter units: '))
if units <= 100:
  print(f'Your electricity bill is Rs.{units * 5}')
elif units > 100 and units <= 200:
 print(f'Your electricity bill is Rs.{units * 7}')
elif units > 200:
 print(f'Your electricity bill is Rs.{units * 10}')
else:
  print("Invalid input")
```

# Q10. Build a mini ATM withdrawal system.

 Take:
 - account balance
 - withdrawal amount

 Conditions:
 - withdrawal amount should not exceed balance
 - withdrawal amount should be multiple of 100

 Print:
 - transaction success
 - remaining balance
 - error message if invalid

```python
balance = float(input("Enter account balance: "))
withdrawal = float(input("Enter withdrawal amount: "))

if withdrawal <= balance and withdrawal % 100 == 0:
    remaining_balance = balance - withdrawal

    print("Transaction Successful!")
    print("Remaining Balance:", remaining_balance)

else:
    print("Transaction Failed!")

    if withdrawal > balance:
        print("Error: Insufficient balance.")

    if withdrawal % 100 != 0:
        print("Error: Withdrawal amount should be multiple of 100.")
```
