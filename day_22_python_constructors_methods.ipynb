# Q1. Create a Student class.

# Q3. Create a BankAccount class.

 Constructor should initialize:
 - account_number
 - account_holder
 - balance

 Create methods:
 - deposit()
 - withdraw()
 - check_balance()

```python
class BankAccount:
    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.balance)
```

```python
acc1 = BankAccount(10101, "Ritesh", 10000)
```

```python
acc1.deposit(5000)
```

```python
acc1.withdraw(3000)
```

```python
acc1.check_balance()
```

# Q4. Create a Product class.

 Constructor should initialize:
 - product_id
 - product_name
 - price

 Create methods:
 - apply_discount()
 - display_product()


```python
class Product:
  def __init__(self,product_id,product_name,price):
    self.product_id = product_id
    self.product_name = product_name
    self.price = price

  def apply_discount(self, percentage):
    discount = self.price * percentage / 100
    self.price -= discount
    print("Discount applied successfully")

  def display_product(self):
    print(f"product_id: {self.product_id}\nproduct_name: {self.product_name}\nprice: {self.price}")
```

```python
p1 = Product(1, "Laptop", 50000)
p1.apply_discount(10)
```

```python
p1.display_product()
```

# Q5. Create a Rectangle class.

 Constructor should initialize:
 - length
 - width

 Create methods:
 - area()
 - perimeter()


```python
class Rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 * (self.length + self.width)
```

```python
r1 = Rectangle(10, 5)
```

```python
print("Area:", r1.area())
print("Perimeter:", r1.perimeter())
```

# Q7. Create a Circle class.

 Constructor should initialize:
 - radius

 Create methods:
 - area()
 - circumference()

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def circumference(self):
        return 2 * 3.14 * self.radius
```

```python
c1 = Circle(7)
```

```python
print("Area:", c1.area())
print("Circumference:", c1.circumference())
```

# Q9. Create a LibraryMember class.

 Constructor should initialize:
 - member_id
 - member_name
 - books_issued

 Create methods:
 - issue_book()
 - return_book()
 - display_status()

```python
class LibraryMember:
    def __init__(self, member_id, member_name):
        self.member_id = member_id
        self.member_name = member_name
        self.books_issued = 0

    def issue_book(self):
        self.books_issued += 1
        print("Book Issued Successfully")

    def return_book(self):
        if self.books_issued > 0:
            self.books_issued -= 1
            print("Book Returned Successfully")
        else:
            print("No Books to Return")

    def display_status(self):
        print("Member ID:", self.member_id)
        print("Member Name:", self.member_name)
        print("Books Issued:", self.books_issued)
```

```python
member1 = LibraryMember(101, "Ritesh")
```

```python
member1.issue_book()
member1.issue_book()
```

```python
member1.display_status()
```

```python
member1.return_book()
```

# Q10. Build a Student Management System.

 Create a Student class.

 Constructor should initialize:
 - student_id
 - name
 - marks

 Create methods:
 - display_details()
 - update_marks()
 - calculate_grade()

 Store all students as objects and
 generate a complete report.

```python
class Student:
    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks

    def display_details(self):
        print(f"ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Grade: {self.calculate_grade()}")

    def update_marks(self, marks):
        self.marks = marks
        print("Marks Updated Successfully")

    def calculate_grade(self):
        if self.marks >= 90:
            return "A"

        elif self.marks >= 75:
            return "B"

        elif self.marks >= 60:
            return "C"

        elif self.marks >= 40:
            return "D"

        else:
            return "F"


students = []


def add_student():
    student_id = int(input("Enter Student ID: "))
    name = input("Enter Name: ")
    marks = float(input("Enter Marks: "))

    student = Student(student_id, name, marks)
    students.append(student)

    print("Student Added Successfully")


def display_students():
    if len(students) == 0:
        print("No Students Found")

    else:
        for student in students:
            print("-" * 30)
            student.display_details()


def update_student_marks():
    student_id = int(input("Enter Student ID: "))

    for student in students:
        if student.student_id == student_id:
            new_marks = float(input("Enter New Marks: "))
            student.update_marks(new_marks)
            return

    print("Student Not Found")


def generate_report():
    if len(students) == 0:
        print("No Students Found")
        return

    print("\nSTUDENT REPORT")
    print("=" * 40)

    for student in students:
        print(
            f"ID: {student.student_id} | "
            f"Name: {student.name} | "
            f"Marks: {student.marks} | "
            f"Grade: {student.calculate_grade()}"
        )


while True:
    print("\n1. Add Student")
    print("2. Display Students")
    print("3. Update Marks")
    print("4. Generate Report")
    print("5. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        add_student()

    elif choice == 2:
        display_students()

    elif choice == 3:
        update_student_marks()

    elif choice == 4:
        generate_report()

    elif choice == 5:
        print("Exiting...")
        break

    else:
        print("Invalid Choice")
```
