# Q3. Create a function that accepts a string and returns the total number of vowels.


```python
def vowels(word):
  count = 0
  for i in word:
    if i in 'AEIOUaeiou':
      count += 1
  return f'Total number of vowels: {count}'
```

```python
vowels('function')
```

# Q5. Create a function that accepts a number and returns its factorial.




```python
def factorial(num):
  fact = 1
  for i in range(1,num+1):
    fact *= i

  return ('Factorial of number is:',fact)
```

```python
factorial(5)
```

# Q6. Create a function that accepts a number and returns whether it is prime or not.


```python
def prime(num):
    if num <= 1:
        return "Number is not prime"

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return "Number is not prime"

    return "Number is prime"
```

```python
prime(46)
```

# Q7. Create a function that accepts marks of three subjects and returns:
 - total marks
 - average marks
 - percentage

```python
def marks(sub1,sub2,sub3):
  total = sub1+sub2+sub3
  return {
      'Total marks':total,
      'Average marks':total/3,
      'Percentage':(total/300)*100
  }
```

```python
marks(76,80,90)
```

# Q9. Create a function that accepts a list of numbers and returns:
 - sum
 - average
 - largest number
 - smallest number

```python
def agger(list1):
  return {
      'Sum':sum(list1),
      'Average':sum(list1)/len(list1),
      'Largest number':max(list1),
      "Smallest number":min(list1)
  }
```

# Q10. Build a simple Student Report Generator.

 Create functions for:
 - input collection
 - total marks calculation
 - average calculation
 - report display

 Generate a formatted student report.

```python
def get_student_data():

    name = input("Enter Student Name: ")
    mark1 = float(input("Enter Subject 1 Marks: "))
    mark2 = float(input("Enter Subject 2 Marks: "))
    mark3 = float(input("Enter Subject 3 Marks: "))

    return name, mark1, mark2, mark3


def calculate_total(mark1, mark2, mark3):

    return mark1 + mark2 + mark3


def calculate_average(total):

    return total / 3


def display_report(name, total, average):

    print("\n----- STUDENT REPORT -----")
    print("Name:", name)
    print("Total Marks:", total)
    print("Average Marks:", round(average, 2))

```

```python

name, mark1, mark2, mark3 = get_student_data()
```

```python

total = calculate_total(mark1, mark2, mark3)
```

```python
average = calculate_average(total)
```

```python

display_report(name, total, average)
```
