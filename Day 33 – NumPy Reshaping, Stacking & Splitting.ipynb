 # Q1. Create a NumPy array containing
 numbers from 1 to 12.

 Reshape it into:

 - 3x4 matrix
 - 4x3 matrix
 - 2x6 matrix

 Display all results.

```python
import numpy as np

arr = np.arange(1,13)

print('3X4 Matrix:\n',arr.reshape(3,4))
print('\n4X3 Matrix:\n',arr.reshape(4,3))
print('\n2X6 Matrix:\n',arr.reshape(2,6))
```

 - 2D array

 - vertical stacking
 - horizontal stacking

 Display results.

# Q6. Create three arrays.

 Combine them using:

 - vstack()
 - hstack()
 - concatenate()

 Compare outputs.

```python

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr3 = np.array([7, 8, 9])

print("Array 1:", arr1)
print("Array 2:", arr2)
print("Array 3:", arr3)

vertical = np.vstack((arr1, arr2, arr3))
print("\nUsing vstack():")
print(vertical)


horizontal = np.hstack((arr1, arr2, arr3))
print("\nUsing hstack():")
print(horizontal)

concat = np.concatenate((arr1, arr2, arr3))
print("\nUsing concatenate():")
print(concat)
```

 # Q9. Create a sales dataset.

 Store sales for:

 - 4 products
 - 6 months

 Generate:

 - product-wise totals
 - month-wise totals

```python
sales = np.array([
    [100, 120, 110, 130, 140, 150],
    [200, 210, 220, 230, 240, 250],
    [150, 160, 170, 180, 190, 200],
    [300, 320, 310, 330, 340, 350]
])

print("Sales Dataset:")
print(sales)
product_totals = np.sum(sales, axis=1)

month_totals = np.sum(sales, axis=0)

print("\nProduct-wise Totals:")
print(product_totals)

print("\nMonth-wise Totals:")
print(month_totals)
```

# Q10. Build a Company Performance Report.

 Create datasets for:

 - sales
 - expenses
 - profit

 Use:

 - reshape()
 - transpose()
 - stacking

 Generate:

 - department-wise analysis
 - monthly analysis
 - overall summary report

```python
import numpy as np

sales = np.array([
    100000, 120000, 110000,
    150000, 160000, 170000,
    90000, 95000, 100000,
    200000, 210000, 220000
]).reshape(4, 3)

expenses = np.array([
    60000, 70000, 65000,
    90000, 95000, 100000,
    50000, 55000, 60000,
    120000, 125000, 130000
]).reshape(4, 3)

profit = sales - expenses

print("=" * 50)
print("COMPANY PERFORMANCE REPORT")
print("=" * 50)

print("\nSales Dataset")
print(sales)

print("\nExpenses Dataset")
print(expenses)

print("\nProfit Dataset")
print(profit)

print("\nSales Transpose")
print(sales.T)

combined = np.stack((sales, expenses, profit))

print("\nCombined Dataset Shape:")
print(combined.shape)

dept_sales = np.sum(sales, axis=1)
dept_expenses = np.sum(expenses, axis=1)
dept_profit = np.sum(profit, axis=1)

print("\nDEPARTMENT-WISE ANALYSIS")

for i in range(4):
    print(
        f"Department {i+1} | "
        f"Sales: {dept_sales[i]} | "
        f"Expenses: {dept_expenses[i]} | "
        f"Profit: {dept_profit[i]}"
    )

month_sales = np.sum(sales, axis=0)
month_expenses = np.sum(expenses, axis=0)
month_profit = np.sum(profit, axis=0)

print("\nMONTH-WISE ANALYSIS")

for i in range(3):
    print(
        f"Month {i+1} | "
        f"Sales: {month_sales[i]} | "
        f"Expenses: {month_expenses[i]} | "
        f"Profit: {month_profit[i]}"
    )

total_sales = np.sum(sales)
total_expenses = np.sum(expenses)
total_profit = np.sum(profit)

print("\nOVERALL SUMMARY REPORT")

print("Total Sales    :", total_sales)
print("Total Expenses :", total_expenses)
print("Total Profit   :", total_profit)

print("Average Sales  :", np.mean(sales))
print("Average Profit :", np.mean(profit))

print("=" * 50)
```
