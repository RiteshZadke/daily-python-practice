 Q1. Tight Coupling Detection

     You are given two classes where one directly creates
     and controls the other.

     Task:
     - Explain why this is tight coupling
     - Refactor to reduce dependency

```python
class Database:
    def connect(self):
        return "Connected to MySQL"

class UserService:
    def __init__(self):
        self.db = Database()

    def process(self):
        return self.db.connect()

# - UserService is responsible for creating Database
# - If we change Database (MySQL → PostgreSQL → API), UserService breaks
# - Hard to test (cannot mock Database easily)
# - Violates Single Responsibility Principle
```

```python
class Database:
    def connect(self):
        return "Connected to MySQL"


class UserService:
    def __init__(self, database):
        self.database = database

    def process(self):
        return self.database.connect()

# WHY this improves flexibility:
# - UserService does NOT care how Database is implemented
# - Easy to swap database (FakeDatabase, TestDatabase, API)
# - Testing becomes trivial
# - Loose coupling = extensible system
```

```python
db = Database()
service = UserService(db)
print(service.process())
```

 Q2. Interface Thinking (Without abc Module)

     Design a system where different services
     (EmailService, SMSService, PushService)
     all expose a send(message) behavior.

     Task:
     - Design a common interface by convention
     - Use polymorphism without forcing inheritance


```python
class EmailService:
    def send(self, message):
        print(f"Email sent: {message}")


class SMSService:
    def send(self, message):
        print(f"SMS sent: {message}")


class PushService:
    def send(self, message):
        print(f"Push notification sent: {message}")
```

```python
class NotificationManager:
    def notify(self, service, message):
        service.send(message)
```

```python
manager = NotificationManager()
```

```python
manager.notify(EmailService(), "Hello")
```

```python
manager.notify(SMSService(), "Hi")
```

```python
manager.notify(PushService(), "Welcome")
```

```python
# WHY this is interview-acceptable:
# - Python is dynamically typed
# - Interface = behavior, not inheritance
# - This is called "Duck Typing"
# - Avoids forced inheritance trees
# - Clean, flexible, Pythonic design
```

 Q3. Avoiding Over-Inheritance

     Scenario:
     - You are tempted to create a deep inheritance tree

     Task:
     - Design the same behavior using composition



```python
class Engine:
    def start(self):
        return "Engine started"


class Battery:
    def charge(self):
        return "Battery charging"


class ElectricCar:
    def __init__(self, engine, battery):
        self.engine = engine
        self.battery = battery

    def drive(self):
        return f"{self.engine.start()} and {self.battery.charge()}"
```

```python
car = ElectricCar(Engine(), Battery())
print(car.drive())
```

```python
# WHY composition is better:
# - Avoids deep inheritance chains
# - Behavior is assembled, not inherited
# - Changes in one component do not break hierarchy
# - More reusable and testable components
```

 Q4. State vs Behavior Separation

     Design a class that performs calculations
     based on input data.

     Task:
     - Separate state (data storage) from behavior
     - Avoid storing unnecessary intermediate state


```python
class Calculator:
    def __init__(self):
        self.last_result = None

    def add(self, a, b):
        self.last_result = a + b
        return self.last_result
```

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b
```

```python
# WHY this improves predictability:
# - No hidden state
# - Same input → same output
# - Easier debugging and testing
# - Safe to reuse in multi-step calculations
```

 Q5. Code Smell Refactoring

     You are given a class that:
     - Has very long methods
     - Uses vague method names
     - Mixes multiple responsibilities

     Task:
     - Identify at least 3 design smells
     - Refactor the class structure
     - Explain each improvement decision in comments

```python
class Report:
    def process(self, data):
        total = 0
        for x in data:
            total += x

        avg = total / len(data)

        print("Total:", total)
        print("Average:", avg)
```

```python
class DataCalculator:
    def calculate_total(self, data):
        return sum(data)

    def calculate_average(self, data):
        return sum(data) / len(data)


class ReportPrinter:
    def print_summary(self, total, average):
        print("Total:", total)
        print("Average:", average)
```

```python
class Report:
    def generate(self, data):
        calculator = DataCalculator()
        printer = ReportPrinter()

        total = calculator.calculate_total(data)
        average = calculator.calculate_average(data)

        printer.print_summary(total, average)
```

```python
# Improvements Explained:
# 1. Long method split into small, focused methods
# 2. Clear, intention-revealing names
# 3. Single Responsibility:
#    - Calculation logic
#    - Presentation logic
#    - Coordination logic
# 4. Easier to extend (CSV, PDF, API output later)
```
