# Q5. Create a BankAccount class.

 Store:
 - account_number
 - account_holder
 - balance

 Create methods:
 - deposit()
 - withdraw()
 - check_balance()

```python
class BankAccount:
    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.balance)
```

```python
acc1 = BankAccount(10101, "Ritesh", 10000)
```

```python
acc1.deposit(5000)
```

```python
acc1.withdraw(3000)
```

```python
acc1.check_balance()
```

# Q6. Create a Rectangle class.

 Store:
 - length
 - width

 Create methods:
 - area()
 - perimeter()


```python
class Rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 * (self.length + self.width)
```

```python
r1 = Rectangle(10, 5)
```

```python
print("Area:", r1.area())
print("Perimeter:", r1.perimeter())
```

# Q7. Create a Product class.

 Store:
 - product_id
 - product_name
 - price

 Create a method to apply
 a percentage discount.

```python
class Product:
  def __init__(self,product_id,product_name,price):
    self.product_id = product_id
    self.product_name = product_name
    self.price = price

  def apply_discount(self, percentage):
    discount = self.price * percentage / 100
    self.price -= discount
    print("Discount applied successfully")

  def display(self):
    print(f"product_id: {self.product_id}\nproduct_name: {self.product_name}\nprice: {self.price}")
```

```python
p1 = Product(1, "Laptop", 50000)
p1.apply_discount(10)
```

```python
p1.display()
```

# Q8. Create a Circle class.

 Store:
 - radius

 Create methods:
 - area()
 - circumference()

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def circumference(self):
        return 2 * 3.14 * self.radius
```

```python
c1 = Circle(7)
```

```python
print("Area:", c1.area())
print("Circumference:", c1.circumference())
```

# Q9. Create a Movie class.

 Store:
 - movie_name
 - director
 - rating

 Create multiple movie objects
 and display highest rated movie.

```python
class Movie:
    def __init__(self, movie_name, director, rating):
        self.movie_name = movie_name
        self.director = director
        self.rating = rating

    def display(self):
        print(self.movie_name, self.director, self.rating)
```

```python
m1 = Movie("Movie A", "Director A", 8.2)
m2 = Movie("Movie B", "Director B", 9.1)
m3 = Movie("Movie C", "Director C", 8.7)
```

```python
movies = [m1, m2, m3]
highest = movies[0]
```

```python
for movie in movies:
    if movie.rating > highest.rating:
        highest = movie

print("Highest Rated Movie:")
highest.display()
```

# Q10. Build a Student Management System.

 Create a Student class.

 Features:
 - add student
 - display students
 - search student
 - update marks

 Store all students as objects.

```python
class Student:
    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks

    def display(self):
        print(self.student_id, self.name, self.marks)


students = []


def add_student():
    student_id = int(input("Enter ID: "))
    name = input("Enter Name: ")
    marks = float(input("Enter Marks: "))

    student = Student(student_id, name, marks)
    students.append(student)

    print("Student Added Successfully")


def display_students():
    if len(students) == 0:
        print("No Students Found")
    else:
        for student in students:
            student.display()


def search_student():
    sid = int(input("Enter Student ID: "))

    for student in students:
        if student.student_id == sid:
            student.display()
            return

    print("Student Not Found")


def update_marks():
    sid = int(input("Enter Student ID: "))

    for student in students:
        if student.student_id == sid:
            new_marks = float(input("Enter New Marks: "))
            student.marks = new_marks
            print("Marks Updated")
            return

    print("Student Not Found")


while True:
    print("\n1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        add_student()

    elif choice == 2:
        display_students()

    elif choice == 3:
        search_student()

    elif choice == 4:
        update_marks()

    elif choice == 5:
        print("Exiting...")
        break

    else:
        print("Invalid Choice")
```
