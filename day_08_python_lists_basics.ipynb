# Q4. Take 8 numbers from the user and store them in a list.
 Count how many numbers are:
 - even
 - odd




```python
l1 = []
even = 0
odd = 0
for i in range(1,9):
  l1.append(int(input(f"Enter number: ")))
for num in l1:
  if num%2 == 0:
    even += 1
  else:
    odd += 1
print('==================================')
print("Even Numbers Count:", even)
print("Odd Numbers Count:", odd)
```

# Q5. Take a list of numbers from the user. Create a new list containing squares of all numbers.

 Example:
 Input:
 [2,3,4]

 Output:
 [4,9,16]

```python
n = int(input("How many numbers? "))

numbers = []

for i in range(n):
    num = int(input(f"Enter number : "))
    numbers.append(num)

squares = []

for num in numbers:
    squares.append(num ** 2)

print("Original List:", numbers)
print("Squares List:", squares)
```

# Q6. Take 10 numbers from the user. Create two separate lists:
 - positive numbers

```python
l1 = []
positive = []
negative = []
for _ in range(1,11):
  l1.append(int(input(f"Enter number: ")))
for i in l1:
  if i > 0:
    positive.append(i)
  else:
    negative.append(i)

print("positive numbers:",positive)
print('negative numbers:',negative)
```

# Q7. Take a list of numbers from the user. Reverse the list without using reverse() method.

```python
n = int(input("How many numbers? "))

numbers = []
rev = []
for i in range(n):
    num = int(input(f"Enter number : "))
    numbers.append(num)
for i in numbers:
  rev.insert(0,i)

print('Reverse:',rev)
```

# Q8. Take 10 numbers from the user. Find the second largest number in the list.

```python
numbers = []

for i in range(10):
    num = int(input(f"Enter number {i+1}: "))
    numbers.append(num)

largest = second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print("Second Largest Number:", second_largest)
```

# Q9. Take a list of numbers from the user. Remove all duplicate values manually.

 Example:
 Input:
 [1,2,2,3,4,4,5]

 Output:
 [1,2,3,4,5]

```python
n = int(input("How many numbers? "))

numbers = []
cleaned = []
for i in range(n):
    num = int(input(f"Enter number : "))
    numbers.append(num)
for i in numbers:
  if i not in cleaned:
    cleaned.append(i)

print(cleaned)
```

# Q10. Build a simple student marks manager. Take marks of 5 students and store them in a list.

 Print:
 - total marks
 - average marks
 - highest marks
 - lowest marks
 - all marks greater than average

```python
marks = []

for i in range(5):
    mark = float(input(f"Enter marks of student {i+1}: "))
    marks.append(mark)

total = sum(marks)
average = total / len(marks)

print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))

print("Marks Greater Than Average:")

for mark in marks:
    if mark > average:
        print(mark)
```
