# Q3. Create two sets.
 Perform:
 - union
 - intersection
 - difference
 - symmetric difference

 Print all results.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
print('Union:',set1.union(set2))
print("Intersection:",set1.intersection(set2))
print("Difference:",set1.difference(set2))
print("Symmetric difference",set1.symmetric_difference(set2))
```

# Q5. Create a set of student names.
 Allow user to:
 - add a new student
 - remove a student
 - check whether a student exists


```python
students = {"Ritesh", "Amit", "Priya"}

new_student = input("Enter student name to add: ")
students.add(new_student)

print("After Adding:", students)

remove_student = input("Enter student name to remove: ")

if remove_student in students:
    students.remove(remove_student)
    print("Student Removed")
else:
    print("Student Not Found")

search_student = input("Enter student name to search: ")

if search_student in students:
    print("Student Exists")
else:
    print("Student Does Not Exist")
```

# Q7. Create a dictionary containing student marks.

 Example:
 {
     "Ritesh": 85,
     "Amit": 78,
     "Priya": 92
 }

 Print:
 - highest marks
 - lowest marks
 - average marks




```python
marks = {
    "Ritesh": 85,
    "Amit": 78,
    "Priya": 92,
    "Neha": 88
}

highest = max(marks.values())
lowest = min(marks.values())

average = sum(marks.values()) / len(marks)

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Average Marks:", round(average, 2))
```

# Q8. Count frequency of each character in a string using a dictionary.

 Example:


 Input:
 hello

 Output:
 {'h':1,'e':1,'l':2,'o':1}




```python
text = input("Enter a string: ")

frequency = {}

for char in text:

    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

# Q10. Build a mini student record system using dictionaries.

 Store:
 - student_id
 - name
 - marks

 Print:
 - all student records
 - topper details
 - average marks

```python
students = [
    {
        "student_id": 101,
        "name": "Ritesh",
        "marks": 85
    },
    {
        "student_id": 102,
        "name": "Amit",
        "marks": 72
    },
    {
        "student_id": 103,
        "name": "Priya",
        "marks": 95
    },
    {
        "student_id": 104,
        "name": "Neha",
        "marks": 88
    }
]

print("All Student Records")

for student in students:
    print(student)

topper = students[0]

for student in students:
    if student["marks"] > topper["marks"]:
        topper = student

total_marks = 0

for student in students:
    total_marks += student["marks"]

average_marks = total_marks / len(students)

print("\nTopper Details")
print(topper)

print("\nAverage Marks")
print(round(average_marks, 2))
```
