# Q1. Take two integers from the user and print:
- Addition
- Subtraction
- Multiplication
- Division
- Floor Division
- Modulus



```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

print(f'Addition: {num1+num2}')
print(f'Subtraction: {num1-num2}')
print(f'Multiplication: {num1*num2}')
print(f'Division: {num1/num2}')
print(f'Floor Division: {num1//num2}')
print(f'Modulus: {num1%num2}')
```

# Q3.Take length and breadth as input from the user.
 Calculate and print:
 - Area of rectangle
- Perimeter of rectangle

```python
length = int(input('Enter the length of rectangle:'))
breadth = int(input('Enter the breadth of rectangle:'))

area = length * breadth
perimeter = 2*(length + breadth)

print(f'Area of rectangle: {area}')
print(f'Perimeter of rectangle: {perimeter}')
```

# Q4.Take radius as input from the user.
Calculate:
- Area of circle
- Circumference of circle

Use:
pi = 3.14


```python
radius = 4
pi = 3.14
area = pi*radius**2
circumference = 2*pi*radius
print(f'Area of circle: {area}')
print(f'Circumference of circle: {circumference}')
```


# Q5.Take temperature in Celsius from the user.Convert it into Fahrenheit.
Formula:
Fahrenheit = (Celsius * 9/5) + 32


```python
celsius = int(input('Enter the temperature in celsius: '))
fahrenheit = (celsius*9/5) + 32
print(f'Temperature in Fahrenheit: {fahrenheit}')
```

# Q6.Take principal amount, rate, and time from user.Calculate Simple Interest.

Formula:
- SI = (P * R * T) / 100

```python
p = int(input('Enter the principal amount: '))
r = int(input('Enter the rate of interest: '))
t = int(input('Enter the tiem: '))

si = (p * r * t) / 100
print(f'Simple interest: {si}')
```


# Q8. Take a character from the user.
Print:
- ASCII value of the character

Then take an ASCII value from user
and convert it back into character.


```python
character = input("Enter any character: ")
ascii = ord(character)
print(f'Character {character} its ascii value {ascii}')

print('=========================================================================')

ascii_value = int(input('Enter ascii value: '))
ch = chr(ascii_value)
print(f'Ascii value {ascii_value} its Character {ch}')
```

# Q9. Take marks of 5 subjects from the user.
Calculate:
- Total marks
- Average marks
- Percentage


```python
eng = int(input('Enter marks of English: '))
math = int(input('Enter marks of Math: '))
sci = int(input('Enter marks of Science: '))
phy = int(input('Enter marks of Physics: '))
chi = int(input('Enter marks of Chemistry: '))

total = eng+math+sci+phy+chi
avg_mark = total/5
per = (total/500)*100
print('=========================================================================')
print(f'Total marks: {total}')
print(f'Average marks: {avg_mark}')
print(f'Percentage: {per}')
```

Q10. Build a mini student report card system.

Print a formatted report card.
