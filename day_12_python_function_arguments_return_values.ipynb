# Q4. Create a function that accepts a string.
 Return:
 - total vowels
 - total consonants
 - total digits


```python
def vowel_cons(word):
  vowel = 0
  conso = 0
  digit = 0
  for i in word:
    if i in 'AEIOUaeiou':
      vowel += 1
    elif i.isalpha():
      conso += 1
    elif i.isdigit():
      digit += 1
  return vowel,conso,digit
```

```python
word = input("Enter a string: ")
result = vowel_cons(word)
print('total vowels:',result[0])
print('total consonants:',result[1])
print('total digits:',result[2])
```

# Q5. Create a function that accepts a number. Return whether the number is:
 - prime
 - even
 - palindrome



```python
def number_properties(number):
  # prime check
  is_prime = True

  if number <= 1:
    is_prime = False

  for i in range(2,int(number ** 0.5)+1):
    if number%i == 0:
      is_prime = False
      break

  # even check
  is_even = number%2 == 0

  # Palindrome
  palindrome = str(number) == str(number)[::-1]

  return is_prime,is_even,palindrome
```

```python
num = int(input("Enter a number: "))

result = number_properties(num)

print("Prime:", result[0])
print("Even:", result[1])
print("Palindrome:", result[2])
```

# Q7. Create a function that accepts multiple student marks.
 Return:
 - highest marks
 - lowest marks
 - average marks


```python
def marks_analysis(*marks):

    highest = max(marks)
    lowest = min(marks)
    average = sum(marks) / len(marks)

    return highest, lowest, average
```

```python
result = marks_analysis(85, 90, 78, 92, 88)

print("Highest Marks:", result[0])
print("Lowest Marks:", result[1])
print("Average Marks:", round(result[2], 2))
```

# Q8. Create a function that accepts a sentence.
 Return:
 - reversed sentence
 - word count
 - longest word


```python
def sentence(sen):
  word_count = len(sen.split())
  longest_word = max(sen.split(),key=len)
  rev_sen = sen[::-1]

  return word_count,longest_word,rev_sen
```

```python
text = input("Enter a sentence: ")

result = sentence(text)

print("Reversed Sentence:", result[0])
print("Word Count:", result[1])
print("Longest Word:", result[2])
```

# Q10. Build a Student Performance Analyzer.

 Create functions for:
 - taking student input
 - calculating total
 - calculating percentage
 - assigning grade
 - displaying final report

 Return values properly between functions.

```python
def get_student_input():

    name = input("Enter Student Name: ")

    mark1 = float(input("Enter Subject 1 Marks: "))
    mark2 = float(input("Enter Subject 2 Marks: "))
    mark3 = float(input("Enter Subject 3 Marks: "))

    return name, mark1, mark2, mark3


def calculate_total(mark1, mark2, mark3):

    return mark1 + mark2 + mark3


def calculate_percentage(total):

    return (total / 300) * 100


def assign_grade(percentage):

    if percentage >= 90:
        return "A"

    elif percentage >= 75:
        return "B"

    elif percentage >= 50:
        return "C"

    else:
        return "Fail"


def display_report(name, total, percentage, grade):

    print("\n----- STUDENT REPORT -----")

    print("Name:", name)
    print("Total Marks:", total)
    print("Percentage:", round(percentage, 2))
    print("Grade:", grade)
```

```python
name, mark1, mark2, mark3 = get_student_input()

total = calculate_total(mark1, mark2, mark3)

percentage = calculate_percentage(total)

grade = assign_grade(percentage)

display_report(name, total, percentage, grade)
```
