# Q1. Print numbers from 1 to 20 using a for loop.


```python
for i in range(1,21):
  print(i,end=' ')
```



# Q2. Print all even numbers from 1 to 50 using a for loop.

```python
for i in range(1,51,2):
  print(i+1,end=' ')
```

# Q3. Print the multiplication table of a number entered by the user.

```python
num = int(input('Enter number: '))
for i in range(1,11):
  print(f'{i} X {num} = {i*num}')
```

# Q4. Take a number from the user. Find the sum of first N natural numbers.

```python
num = int(input('Enter number: '))
sum = 0
for i in range(1,num+1):
  sum += i
print("Sum of first N natural numbers: ",sum)
```

# Q5. Take a number from the user. Calculate factorial using a for loop.


```python
num = int(input('Enter number: '))
factorial = 1
for i in range(1,num+1):
  factorial = factorial*i

print('Factorial:',factorial)
```


# Q6. Take a string from the user. Count the number of vowels in the string using a loop.


```python
st = input("Enter string: ")
vowels = 0
for i in st:
  if i in 'AEIOUaeiou':
    vowels += 1
print('Vowel:',vowels)
```

# Q9. Take a number from the user. Check whether the number is prime or not using a loop.




```python
num = int(input('Enter number: '))
count = 0
for i in range(1,num+1):
  if num%i == 0:
    count += 1

if count == 2:
  print('Number is prime')
else:
  print('Number is not prime')
```

# Q10. Take 5 numbers from the user using a loop.
 Find:
 - largest number
 - smallest number
 - total sum
 - average

```python
l1 = []
total = 0
avg = 0
large = float('-inf')
small = float('inf')
for _ in range(1,6):
  l1.append(int(input("Enter number: ")))
for i in l1:
  total += i
  if i > large:
    large = i
for i in l1:
  if i < small:
    small = i
avg = total/5
print("==========================")
print("largest number:",large)
print("smallest number:",small)
print("total sum:",total)
print("average:",avg)
```
