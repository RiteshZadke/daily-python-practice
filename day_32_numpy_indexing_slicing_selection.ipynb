# Q1. Create a NumPy array containing numbers from 1 to 20.

 Display:

 - first element
 - last element
 - fifth element
 - tenth element

```python
import numpy as np

arr = np.arange(1,21)
print('Array:',arr)
print('First Element:',arr[0])
print('Last Element:',arr[-1])
print('Fifth Element:',arr[4])
print('Tenth Element:',arr[9])
```

 # Q2. Create a NumPy array containing numbers from 1 to 30.

 Extract and display:

 - first 10 elements
 - last 10 elements
 - elements from index 5 to 15
 - every second element

```python
arr1 = np.arange(1,31)

print('Array:',arr1)
print('First 10 elements:',arr1[:9])
print('Last 10 elements:',arr1[-10:])
print('Element from index 5 to 15:',arr1[5:14])
print('Every Second element:',arr1[::2])
```

# Q4. Create an array containing numbers from 1 to 50.

 Using boolean indexing display:

 - numbers greater than 25
 - numbers less than 10
 - even numbers only
 - odd numbers only

```python
arr3 = np.arange(1,51)
print('Number greater than 25:',arr3[arr3 > 25])
print('Number less than 10:',arr3[arr3 < 10])
print('Even numbers only:',arr3[arr3%2 == 1])
print('Odd numbers only:',arr3[arr3%2 != 0])
```

# Q5. Create a salary dataset.

 Calculate:

 - average salary

 Display:

 - salaries greater than average salary
 - salaries less than average salary
 - highest salary
 - lowest salary

```python
arr = np.array([45000,55000,50000,48000,40000,60000,65000])

avg = np.sum(arr)/np.size(arr)
print("Average Salary:",avg)
print("Salaries greater than average salary:",arr[arr > avg])
print("Salaries less than average salary:",arr[arr < avg])
print("Highest salary:",np.max(arr))
print("Lowest salary:",np.min(arr))
```

# Q9. Create an array containing 25 random integers.

 Using fancy indexing display:

 - 1st element
 - 5th element
 - 10th element
 - 15th element

 Also display:

 - top 5 values
 - bottom 5 values

```python
import numpy as np

arr = np.random.randint(1, 101, 25)

print("Original Array:")
print(arr)

print("1st Element :", arr[[0]])
print("5th Element :", arr[[4]])
print("10th Element:", arr[[9]])
print("15th Element:", arr[[14]])

sorted_arr = np.sort(arr)

print("\nTop 5 Values:")
print(sorted_arr[-5:])

print("\nBottom 5 Values:")
print(sorted_arr[:5])
```

# Q10. Build a Company Employee Analytics Report.

 Create datasets for:

 - employee IDs
 - salaries
 - years of experience

 Generate:

 - employees with salary above average
 - employees with experience greater than 5 years
 - employees eligible for promotion

 Display a complete analytics report.

```python
import numpy as np

# Employee Dataset
employee_ids = np.array([101, 102, 103, 104, 105, 106, 107, 108])
salaries = np.array([50000, 75000, 45000, 90000, 65000, 55000, 80000, 60000])
experience = np.array([2,7,1,10,6,3,8,5])
avg_salary = np.mean(salaries)
above_avg_mask = salaries > avg_salary
experienced_mask = experience > 5
promotion_mask = (
    (experience > 5) &
    (salaries > avg_salary)
)

print("=" * 50)
print("COMPANY EMPLOYEE ANALYTICS REPORT")
print("=" * 50)
print("\nAverage Salary:", round(avg_salary, 2))
print("\nEmployees With Salary Above Average")
print(employee_ids[above_avg_mask])
print("\nEmployees With Experience > 5 Years")
print(employee_ids[experienced_mask])
print("\nEmployees Eligible For Promotion")
print(employee_ids[promotion_mask])
print("\nPromotion Details")
print("Employee IDs :", employee_ids[promotion_mask])
print("Salaries     :", salaries[promotion_mask])
print("Experience   :", experience[promotion_mask])
print("=" * 50)
```
