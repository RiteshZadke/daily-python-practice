# Q1. Create a Person class.

 Attributes:
 - name
 - age

 Create a Student class that inherits Person.

 Additional Attribute:
 - marks

 Display all details.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)
```

```python
s1 = Student("Ritesh", 22, 88)
```

```python
s1.display()
```

# Q3. Create an Employee class.

 - salary

 Create a Manager class that inherits Employee.

 Additional Attribute:
 - team_size

```python
class Employee:
    def __init__(self,employee_id, employee_name, salary):
        self.employee_id = employee_id
        self.employee_name = employee_name
        self.salary = salary


class Manager(Employee):
    def __init__(self, employee_id, employee_name, salary, team_size):
        super().__init__(employee_id,employee_name, salary)
        self.team_size = team_size

    def display(self):
        print("employee_id",self.employee_id)
        print("employee_name:", self.employee_name)
        print("salary:", self.salary)
        print("team_size:", self.team_size)
```

# Q4. Create a BankAccount class.

 Attributes:
 - account_number
 - balance

 Create a SavingsAccount class.

 Additional Attribute:
 - interest_rate

 Create methods to:
 - calculate interest
 - display account details




```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance


class SavingsAccount(BankAccount):
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate

    def calculate_interest(self):
        return self.balance * self.interest_rate / 100

    def display(self):
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)
        print("Interest Rate:", self.interest_rate)
        print("Interest:", self.calculate_interest())
```

```python
s1 = SavingsAccount(1001, 50000, 5)
```

```python
s1.display()
```

# Q9. Create a CompanyEmployee class.

 Create child classes:
 - Developer
 - Tester
 - Analyst

 Add role-specific methods for each class.

```python
class CompanyEmployee:
    def __init__(self, employee_id, employee_name):
        self.employee_id = employee_id
        self.employee_name = employee_name


class Developer(CompanyEmployee):
    def develop(self):
        print(self.employee_name, "is Writing Code")


class Tester(CompanyEmployee):
    def test(self):
        print(self.employee_name, "is Testing Software")


class Analyst(CompanyEmployee):
    def analyze(self):
        print(self.employee_name, "is Analyzing Requirements")
```

```python
d1 = Developer(1, "Ritesh")
t1 = Tester(2, "Amit")
a1 = Analyst(3, "Neha")
```

```python
d1.develop()
t1.test()
a1.analyze()
```

# Q10. Build a School Management System.

 Parent Class:
 - Person

 Child Classes:
 - Student
 - Teacher

 Store:
 - personal details
 - role-specific details

 Generate a complete school report
 using inheritance.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, student_id, marks):
        super().__init__(name, age)
        self.student_id = student_id
        self.marks = marks

    def display(self):
        print("Student Details")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Student ID:", self.student_id)
        print("Marks:", self.marks)


class Teacher(Person):
    def __init__(self, name, age, teacher_id, subject):
        super().__init__(name, age)
        self.teacher_id = teacher_id
        self.subject = subject

    def display(self):
        print("Teacher Details")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Teacher ID:", self.teacher_id)
        print("Subject:", self.subject)
```

```python
people = []

people.append(Student("Ritesh", 22, 101, 88))
people.append(Student("Amit", 21, 102, 76))

people.append(Teacher("Sharma", 45, 201, "Physics"))
people.append(Teacher("Patil", 38, 202, "Mathematics"))
```

```python
print("SCHOOL REPORT")
print("=" * 30)

for person in people:
    person.display()
    print("-" * 30)
```
