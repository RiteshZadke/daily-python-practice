# Q1. Create a custom iterator that generates numbers from 1 to N.

 Implement:
 - __iter__()
 - __next__()

```python
class NumberIterator:
    def __init__(self, n):
        self.n = n
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.n:
            value = self.current
            self.current += 1
            return value
        raise StopIteration
```

```python
nums = NumberIterator(5)
```

```python
for num in nums:
    print(num)
```

 # Q2. Create a custom iterator that generates even numbers up to N.

```python
class EvenIterator:
    def __init__(self, n):
        self.n = n
        self.current = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.n:
            value = self.current
            self.current += 2
            return value
        raise StopIteration
```

```python
evens = EvenIterator(10)
```

```python
for num in evens:
    print(num)
```

 # Q3. Create a generator function that generates squares of numbers from 1 to N.

 Use yield.

```python
def square_generator(n):
    for i in range(1, n + 1):
        yield i ** 2
```

```python
for square in square_generator(5):
    print(square)
```

 # Q4. Create a generator function that generates Fibonacci numbers.

```python
def fibonacci(n):
    a = 0
    b = 1

    for _ in range(n):
        yield a
        a, b = b, a + b
```

```python
for num in fibonacci(10):
    print(num)
```

 # Q5. Create a generator that reads values one at a time from a list.

```python
def list_generator(data):
    for item in data:
        yield item
```

```python
numbers = [10, 20, 30, 40, 50]
```

```python
for value in list_generator(numbers):
    print(value)
```

# Q6. Create a decorator that displays:

 "Function Started"

 "Function Finished"

 before and after function execution.

```python
def logger(func):

    def wrapper():
        print("Function Started")

        func()

        print("Function Finished")

    return wrapper


@logger
def greet():
    print("Hello World")
```

```python
greet()
```

# Q7. Create a decorator that calculates execution time of a function.

```python
import time

def timer(func):

    def wrapper():
        start = time.time()

        func()

        end = time.time()

        print("Execution Time:", end - start)

    return wrapper


@timer
def task():
    time.sleep(2)
    print("Task Completed")
```

```python
task()
```

 # Q8. Create a decorator that validates positive numeric inputs.

```python
def validate_positive(func):

    def wrapper(num):

        if num <= 0:
            print("Number Must Be Positive")
            return

        return func(num)

    return wrapper


@validate_positive
def square(num):
    print(num ** 2)
```

```python
square(5)
square(-2)
```

 # Q9. Create multiple decorators and apply them to the same function.

```python
def logger(func):

    def wrapper():
        print("Function Started")

        func()

        print("Function Finished")

    return wrapper


def timer(func):

    def wrapper():
        import time

        start = time.time()

        func()

        end = time.time()
        print("Time:", end - start)
    return wrapper


@logger
@timer
def task():
    print("Processing...")
```

```python
task()
```

# Q10. Build a Data Processing Pipeline.

 Use:
 - generators
 - decorators
 - functions

 Process a list of student marks.

 Generate:
 - filtered data
 - transformed data
 - summary report

```python
def report(func):

    def wrapper(data):
        print("Processing Started")

        result = func(data)

        print("Processing Finished")

        return result

    return wrapper


def filter_marks(data):
    for mark in data:
        if mark >= 40:
            yield mark


def add_bonus(data):
    for mark in data:
        yield mark + 5


@report
def process_marks(data):

    filtered = filter_marks(data)
    transformed = add_bonus(filtered)
    result = list(transformed)
    print("Processed Marks:", result)
    print("Average:", sum(result) / len(result))

    return result
```

```python
marks = [25, 45, 60, 80, 30, 90]
```

```python
process_marks(marks)
```
