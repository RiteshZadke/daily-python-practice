# Q4. Create a Product class.

 Class Variable:
 - tax_rate

 Instance Variables:
 - product_id
 - product_name
 - price

 Create a method to calculate
 final price including tax.

```python
class Product:
  tax_rate = 18

  def __init__(self,product_id,name,price):
    self.product_id = product_id
    self.name = name
    self.price = price

  def final_price(self):
    tax = self.price*Product.tax_rate/100
    return self.price + tax

  def display(self):
   print("Product ID:", self.product_id)
   print("Product Name:", self.name)
   print("Price:", self.price)
   print("Final Price:", self.final_price())
```

```python
p1 = Product(101, "Laptop", 50000)
```

```python
p1.display()
```

# Q6. Create a LibraryMember class.

 Class Variable:
 - max_books_allowed

 Instance Variables:
 - member_id
 - member_name
 - books_issued

 Create methods:
 - issue_book()
 - display_status()

```python
class LibraryMember:
    max_books_allowed = 5

    def __init__(self, member_id, member_name):
        self.member_id = member_id
        self.member_name = member_name
        self.books_issued = 0

    def issue_book(self):
        if self.books_issued < LibraryMember.max_books_allowed:
            self.books_issued += 1
            print("Book Issued Successfully")
        else:
            print("Maximum Book Limit Reached")

    def display_status(self):
        print("Member ID:", self.member_id)
        print("Member Name:", self.member_name)
        print("Books Issued:", self.books_issued)
        print("Max Books Allowed:", LibraryMember.max_books_allowed)
```

```python
m1 = LibraryMember(101, "Ritesh")
```

```python
m1.issue_book()
m1.issue_book()
```

```python
m1.display_status()
```

# Q10. Build a Student Management System.

 Class Variable:
 - school_name
 - total_students

 Instance Variables:
 - student_id
 - name
 - marks

 Create methods:
 - display_details()
 - update_marks()
 - calculate_grade()

 Track total students created using
 a class variable.

```python
class Student:
    school_name = "ABC School"
    total_students = 0

    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks

        Student.total_students += 1

    def display_details(self):
        print(f"School: {Student.school_name}")
        print(f"ID: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Grade: {self.calculate_grade()}")

    def update_marks(self, marks):
        self.marks = marks
        print("Marks Updated Successfully")

    def calculate_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 40:
            return "D"
        else:
            return "F"


students = []


def add_student():
    student_id = int(input("Enter Student ID: "))
    name = input("Enter Name: ")
    marks = float(input("Enter Marks: "))

    student = Student(student_id, name, marks)
    students.append(student)

    print("Student Added Successfully")


def display_students():
    if len(students) == 0:
        print("No Students Found")
    else:
        for student in students:
            print("-" * 30)
            student.display_details()

        print("\nTotal Students:", Student.total_students)


while True:
    print("\n1. Add Student")
    print("2. Display Students")
    print("3. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        add_student()

    elif choice == 2:
        display_students()

    elif choice == 3:
        print("Exiting...")
        break

    else:
        print("Invalid Choice")
```
