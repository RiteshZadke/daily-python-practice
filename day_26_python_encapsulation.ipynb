 # Q1. Create a Student class.

 Private Attributes:
 - __student_id
 - __marks

 Create methods:
 - set_marks()
 - get_marks()
 - display_student()

```python
class Student:
  def __init__(self,student_id,marks):
    self.__student_id = student_id
    self.__marks = marks

  def get_marks(self):
    return self.__marks

  def set_marks(self,marks):
    if marks > 0:
      self.__marks = marks
  def display_student(self):
   print("Student ID:", self.__student_id)
   print("Marks:", self.__marks)
```

```python
s1 = Student(101, 85)
```

```python
s1.display_student()
```

```python
s1.set_marks(92)
```

```python
print("Updated Marks:", s1.get_marks())
```

```python
s1.display_student()
```

 # Q2. Create a BankAccount class.

 Private Attribute:
 - __balance

 Create methods:
 - deposit()
 - withdraw()
 - get_balance()

 Prevent invalid withdrawals.

```python
class BankAccount:
  def __init__(self,balance):
    self.__balance = balance

  def deposit(self,amount):
    if amount > 0:
      self.__balance += amount
      print("Deposit Successfull")

    else:
      print("Deposit Amount Should be greater the 0")

  def withdraw(self,amount):
    if amount <= self.__balance:
      self.__balance -= amount
      print('Withdraw Successfull')

    else:
      print("Insufficient Balance")

  def get_balance(self):
    return self.__balance
```

```python
acc = BankAccount(10000)
acc.deposit(5000)
acc.withdraw(3000)
```

```python
print("Balance:", acc.get_balance())
```

 - __books_issued

 - issue_book()
 - return_book()
 - get_books_issued()

 Books issued cannot be negative.

```python
class LibraryMember:
    def __init__(self):
        self.__books_issued = 0

    def issue_book(self):
        self.__books_issued += 1

    def return_book(self):
        if self.__books_issued > 0:
            self.__books_issued -= 1
        else:
            print("No Books To Return")

    def get_books_issued(self):
        return self.__books_issued
```

 Private Attributes:
 - __patient_id

 Create methods:
 - update_record()

 Restrict direct access.

```python
class HospitalPatient:
    def __init__(self, patient_id, medical_record):
        self.__patient_id = patient_id
        self.__medical_record = medical_record

    def update_record(self, new_record):
        self.__medical_record = new_record
        print("Medical Record Updated")

    def view_record(self):
        print("Patient ID:", self.__patient_id)
        print("Medical Record:", self.__medical_record)
```

# Q10. Build a Bank Management System.

 Private Attributes:
 - __account_number
 - __balance

 Features:
 - deposit
 - withdraw
 - check balance
 - transaction history

 Use encapsulation to protect
 all sensitive account data.

```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance
        self.__transaction_history = []

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__transaction_history.append(
                f"Deposited ₹{amount}"
            )

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            self.__transaction_history.append(
                f"Withdrawn ₹{amount}"
            )
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.__balance)

    def show_transactions(self):
        print("\nTransaction History")

        for transaction in self.__transaction_history:
            print(transaction)
```

```python
acc = BankAccount(1001, 10000)
```

```python
acc.deposit(5000)
acc.withdraw(3000)
```

```python
acc.check_balance()
```

```python
acc.show_transactions()
```
