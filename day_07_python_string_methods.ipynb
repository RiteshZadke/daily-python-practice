# Q1. Take a sentence from the user.
 Print:
 - Original sentence
 - Uppercase version
 - Lowercase version
 - Title case version



```python
sen = input("Enter Sentense: ")
print('=======================================')
print('Original sentence:',sen)
print('Uppercase version:',sen.upper())
print('Lowercase version:',sen.lower())
print('Title case version:',sen.title())
```

# Q2. Take a string from the user. Count how many times a specific character appears.

-  Character should also be taken from the user.


```python
sen = input("Enter Sentense: ")
ch = input('Enter Character: ')
print('Character apperenc:',sen.count(ch))
```

# Q5. Take a sentence from the user. Check whether a given word exists inside the sentence.
 - Word should also be taken from the user.


```python
sen = input("Enter a sentence: ")
word = input("Enter a word to search: ")

if word in sen:
    print("Word exists in the sentence.")
else:
    print("Word does not exist in the sentence.")
```

# Q6. Take a full name from the user. Convert the name into initials.

 Example:
- Ritesh Zadke
- Output:
- R.Z


```python
name = input('Enter your full name: ').replace(' ','.')
con = ''
for i in name:
  if i.isupper() or i == '.':
    con += i
print(con)
```

# Q7. Take a sentence from the user.
 Count:
 - Total words
 - Total characters
 - Total spaces

```python
sen = input("Enter a sentence: ")
word = 0
ch = 0
for i in sen.split():
  word += 1
for i in sen:
  if i.isalnum():
    ch += 1

print('Total words:',word)
print('Total characters:',ch)
print('Total spaces:',sen.count(' '))
```

# Q8. Take a string from the user. Remove all vowels from the string.
 Example:
- Input: Programming
- Output: Prgrmmng



```python
sen = input("Enter a string: ")
rem = ''
for i in sen:
  if i not in 'AEIOUaeiou':
    rem += i
print(rem)
```

# Q9. Take a sentence from the user. Reverse the order of words.

 Example:
 Input:
- I Love Python

 Output:
- Python Love I


```python
sen = input("Enter a sentence: ").split()
rev =''
for i in sen:
  rev = i+' '+rev

print(rev)
```



# Q10. Take a sentence from the user.
 Find:
 - longest word
 - shortest word

 Print both words and their lengths.

```python
sen = input("Enter a sentence: ").split()

longest = sen[0]
shortest = sen[0]

for i in sen:
    if len(i) > len(longest):
        longest = i

    if len(i) < len(shortest):
        shortest = i

print("Longest Word:", longest)
print("Shortest Word:", shortest)
```
