 # Q4. Create a Rectangle class.

 Implement:
 - __init__()
 - __str__()
 - __add__()

 Adding two rectangles should return
 the sum of their areas.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def __str__(self):
        return f"Rectangle({self.length}, {self.width})"

    def __add__(self, other):
        return self.area() + other.area()
```

```python
r1 = Rectangle(10, 5)
r2 = Rectangle(4, 5)
```

```python
print(r1)
print(r2)
```

```python
print("Total Area:", r1 + r2)
```

 # Q5. Create a BankAccount class.

 Implement:
 - __init__()
 - __str__()
 - __eq__()

 Compare two accounts based on balance.

```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def __str__(self):
        return (
            f"Account No: {self.account_number}, "
            f"Balance: {self.balance}"
        )

    def __eq__(self, other):
        return self.balance == other.balance
```

```python
acc1 = BankAccount(101, 10000)
acc2 = BankAccount(102, 10000)
acc3 = BankAccount(103, 15000)
```

```python
print(acc1)
```

```python
print(acc1 == acc2)
print(acc1 == acc3)
```

 # Q8. Create a ShoppingCart class.

 Implement:
 - __init__()
 - __len__()
 - __str__()

 Store products and display
 cart information.

```python
class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def __len__(self):
        return len(self.products)

    def __str__(self):
        return f"Shopping Cart: {self.products}"
```

```python
cart = ShoppingCart()
```

```python
cart.add_product("Laptop")
cart.add_product("Mouse")
cart.add_product("Keyboard")
```

```python
print(cart)
print("Total Products:", len(cart))
```

 # Q9. Create a Vector class.

 Implement:
 - __init__()
 - __add__()

 Add two vectors together.

 Example:
 (2,3) + (4,5) = (6,8)

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(
            self.x + other.x,
            self.y + other.y
        )

    def __str__(self):
        return f"({self.x}, {self.y})"
```

```python
v1 = Vector(2, 3)
v2 = Vector(4, 5)
```

```python
v3 = v1 + v2
```

```python
print(v3)
```

# Q10. Build a Student Management System.

 Create a Student class.

 Implement:
 - __init__()
 - __str__()
 - __repr__()
 - __eq__()

 Features:
 - add student
 - compare students
 - display student records

 Use magic methods to make
 the system more user-friendly.

```python
class Student:
    def __init__(self, student_id, name, marks):
        self.student_id = student_id
        self.name = name
        self.marks = marks

    def __str__(self):
        return (
            f"ID: {self.student_id}, "
            f"Name: {self.name}, "
            f"Marks: {self.marks}"
        )

    def __repr__(self):
        return (
            f"Student({self.student_id}, "
            f"'{self.name}', {self.marks})"
        )

    def __eq__(self, other):
        return self.marks == other.marks
```

```python
students = []

students.append(Student(101, "Ritesh", 85))
students.append(Student(102, "Amit", 85))
students.append(Student(103, "Neha", 92))
```

```python
print("Student Records\n")

for student in students:
    print(student)

print("\nComparison")

print(students[0] == students[1])
print(students[0] == students[2])

print("\nRepresentation")

print(repr(students[0]))
```
