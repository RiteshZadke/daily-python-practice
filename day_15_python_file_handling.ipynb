# Q5. Create a program that reads a file and displays only lines containing a specific keyword.

- Keyword should be taken from user input.

```python
target = input("Enter keyword: ")

with open("notes.txt", "r") as file:
    lines = file.readlines()

matching_lines = []

for line in lines:
    if target in line:
        matching_lines.append(line.strip())

if matching_lines:
    print("Matching Lines:")

    for line in matching_lines:
        print(line)
else:
    print("Keyword not found")
```

# Q6. Create a program that copies content from one file into another file.

 Example:
- source.txt → backup.txt

```python
with open("notes.txt",'r') as src, open('backup','w') as backup:
  for line in src:
    backup.write(line)

with open('backup','r') as file:
  content = file.read()

print(content)
```

# Q9. Create a program that reads student records from a file and finds:
 - topper
 - average marks
 - lowest marks

```python
students = []

with open("student_records.txt", "r") as file:
    while True:
        id_line = file.readline()
        if not id_line:  # End of file
            break
        name_line = file.readline()
        marks_line = file.readline()
        separator_line = file.readline() # Read the separator line

        student_id = int(id_line.split(": ")[1].strip())
        name = name_line.split(": ")[1].strip()
        marks = int(marks_line.split(": ")[1].strip())

        students.append({
            "student_id": student_id,
            "name": name,
            "marks": marks
        })


if students:
    topper = max(students, key=lambda student: student["marks"])
    lowest = min(students, key=lambda student: student["marks"])
    average_marks = sum(student["marks"] for student in students) / len(students)

    print("Topper:")
    print(topper)

    print("\nLowest Scorer:")
    print(lowest)

    print("\nAverage Marks:")
    print(round(average_marks, 2))
else:
    print("No student records found.")
```

# Q10. Build a Mini Student File Management System.

 Features:
 - Add student
 - View students
 - Search student
 - Update student marks
 - Delete student

 Store all records in a text file.

```python
FILE_NAME = "students.txt"


def add_student():

    student_id = input("Enter Student ID: ")
    name = input("Enter Student Name: ")
    marks = input("Enter Student Marks: ")

    with open(FILE_NAME, "a") as file:
        file.write(f"{student_id},{name},{marks}\n")

    print("Student Added Successfully")


def view_students():

    try:
        with open(FILE_NAME, "r") as file:

            records = file.readlines()

            if not records:
                print("No Records Found")
                return

            print("\nStudent Records")

            for record in records:
                student_id, name, marks = record.strip().split(",")

                print(
                    "ID:", student_id,
                    "| Name:", name,
                    "| Marks:", marks
                )

    except FileNotFoundError:
        print("No Student File Found")


def search_student():

    target_id = input("Enter Student ID to Search: ")

    found = False

    try:
        with open(FILE_NAME, "r") as file:

            for record in file:

                student_id, name, marks = record.strip().split(",")

                if student_id == target_id:

                    print("\nStudent Found")
                    print("ID:", student_id)
                    print("Name:", name)
                    print("Marks:", marks)

                    found = True
                    break

        if not found:
            print("Student Not Found")

    except FileNotFoundError:
        print("No Student File Found")


def update_marks():

    target_id = input("Enter Student ID: ")

    updated_records = []

    found = False

    try:
        with open(FILE_NAME, "r") as file:

            for record in file:

                student_id, name, marks = record.strip().split(",")

                if student_id == target_id:

                    new_marks = input("Enter New Marks: ")

                    updated_records.append(
                        f"{student_id},{name},{new_marks}\n"
                    )

                    found = True

                else:
                    updated_records.append(record)

        with open(FILE_NAME, "w") as file:
            file.writelines(updated_records)

        if found:
            print("Marks Updated Successfully")
        else:
            print("Student Not Found")

    except FileNotFoundError:
        print("No Student File Found")


def delete_student():

    target_id = input("Enter Student ID: ")

    updated_records = []

    found = False

    try:
        with open(FILE_NAME, "r") as file:

            for record in file:

                student_id, name, marks = record.strip().split(",")

                if student_id == target_id:
                    found = True

                else:
                    updated_records.append(record)

        with open(FILE_NAME, "w") as file:
            file.writelines(updated_records)

        if found:
            print("Student Deleted Successfully")
        else:
            print("Student Not Found")

    except FileNotFoundError:
        print("No Student File Found")


while True:

    print("\n===== STUDENT FILE MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Student Marks")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        view_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        update_marks()

    elif choice == "5":
        delete_student()

    elif choice == "6":
        print("Program Closed")
        break

    else:
        print("Invalid Choice")
```
