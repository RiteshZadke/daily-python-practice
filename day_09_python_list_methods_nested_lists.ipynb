# Q1. Create a list of 10 numbers.

 - append()

# Q2. Create a list of numbers.
 Remove elements using:
 - remove()
 - pop()
 - clear()

 Explain the difference through comments.


```python
nums = [1,2,5,9,11,17,22,40,15,19]
nums.remove(5)
print(nums)
# remove() remove the element that we have specified from the list.
```

```python
nums = [1,2,5,9,11,17,22,40,15,19]
nums.pop(9)
print(nums)
# pop() method removes the element from the list based on index value and displays that element after pop.
```

```python
nums = [1,2,5,9,11,17,22,40,15,19]
nums.clear()
print(nums)
# clear method make list empty by removing all the elements from the list.
```

# Q3. Take a list of numbers from the user.
 Sort the list:
 - ascending
 - descending

 Print both results.


```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

ascending = numbers.copy()
ascending.sort()

descending = numbers.copy()
descending.sort(reverse=True)

print("Ascending:", ascending)
print("Descending:", descending)
```

# Q4. Take a number from the user. Count how many times it appears in a list.

- Use count() method.


```python
numbers = [10, 20, 10, 30, 40, 10, 50]

target = int(input("Enter number to count: "))

count = numbers.count(target)

print(f"{target} appears {count} times")
```

# Q5. Take a number from the user. Find its first occurrence position in a list.

- Use index() method.

 Handle the case when the number does not exist.




```python
numbers = [10, 20, 30, 40, 20, 50]
target = int(input("Enter number to search: "))

if target in numbers:
  position = numbers.index(target)
  print("First occurrence at index:", position)
else:
    print("Number not found")
```

# Q8. Create a nested list representing a 3x3 matrix.

 Print:
 - each row separately
 - all elements using nested loops




```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print('Rows:')
for row in matrix:
  print(row)

print("\nElements:")
for row in matrix:
  for ele in row:
    print(ele,end=' ')
```

# Q9. Create a nested list containing:
 - student name
 - marks in 3 subjects

 Calculate and print:
 - total marks
 - average marks

 for each student.


```python
students = [
    ["Ritesh", [80, 85, 90]],
    ["Amit", [70, 75, 80]],
    ["Priya", [90, 92, 95]]
]

for student in students:
    name = student[0]
    marks = student[1]

    total = sum(marks)
    average = total / len(marks)

    print("Name:", name)
    print("Total:", total)
    print("Average:", round(average, 2))
    print()
```

# Q10. Build a mini student management system.

 Store records using nested lists.

 Each record should contain:
 - student id
 - student name
 - marks

 Print:
 - all records
 - highest scoring student
 - lowest scoring student
 - average marks of all students

```python
students = [
    [101, "Ritesh", 85],
    [102, "Amit", 70],
    [103, "Priya", 95],
    [104, "Neha", 80]
]

print("All Student Records")

for student in students:
    print(student)

highest = students[0]
lowest = students[0]

total_marks = 0

for student in students:

    total_marks += student[2]

    if student[2] > highest[2]:
        highest = student

    if student[2] < lowest[2]:
        lowest = student

average_marks = total_marks / len(students)

print("\nHighest Scoring Student")
print(highest)

print("\nLowest Scoring Student")
print(lowest)

print("\nAverage Marks")
print(round(average_marks, 2))
```
