# Q1. Create an Animal class.

 - Cat

```python
class Animal:
  def make_sound(self):
    print('Animal makes a sound')

class Dog(Animal):
  def make_sound(self):
    print("Barks")

class Cat(Animal):
  def make_sound(self):
    print("Meows")
```

# Q4. Create a Shape class.

Create child classes:
- Circle
- Rectangle

Create a method:
- calculate_area()

Override the method in each child class.

```python
class Shape:
  def calculate_area(self):
    pass

class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius

  def calculate_area(self):
    return 3.14*self.radius*2

class Rectangle(Shape):
  def __init__(self,length,breath):
    self.length = length
    self.breath = breath

  def calculate_area(self):
    return self.length*self.breath
```

```python
c1 = Circle(7)
r1 = Rectangle(10, 5)
```

```python
c1.calculate_area()
```

```python
r1.calculate_area()
```

# Q10. Build a Company Management System.

 Parent Class:
 - Employee

 Child Classes:
 - Developer
 - Tester
 - Manager

 Create methods:
 - work()
 - calculate_bonus()

 Demonstrate runtime polymorphism by
 storing all objects in a single list
 and calling common methods dynamically.

```python
class Employee:
    def work(self):
        print("Employee is working")

    def calculate_bonus(self):
        print("Bonus Calculated")


class Developer(Employee):
    def work(self):
        print("Developer is writing code")

    def calculate_bonus(self):
        print("Developer Bonus: ₹20,000")


class Tester(Employee):
    def work(self):
        print("Tester is testing applications")

    def calculate_bonus(self):
        print("Tester Bonus: ₹15,000")


class Manager(Employee):
    def work(self):
        print("Manager is managing projects")

    def calculate_bonus(self):
        print("Manager Bonus: ₹30,000")
```

```python
employees = [
    Developer(),
    Tester(),
    Manager()
]

for employee in employees:
    employee.work()
    employee.calculate_bonus()
    print("-" * 30)
```
