# Q1. Print numbers from 1 to 20 using a while loop.

```python
i = 1
while i <= 20:
  print(i,end=' ')
  i += 1
```


# Q2. Print all even numbers from 2 to 50 using a while loop.


```python
i = 1
while i <= 50:
  if i%2 == 0:
    print(i,end=' ')
  i += 1
```

# Q3. Take a number from the user. Print its multiplication table using a while loop.


```python
num = int(input("Enter number: "))
i = 1
while i <= 10:
  print(f'{i} X {num} = {i*num}')
  i += 1
```

# Q4. Take a number N from the user. Find the sum of first N natural numbers using a while loop.

```python
num = int(input("Enter number: "))
i = 1
total = 0
while i <= num:
  total += i
  i += 1
print("Sum :",total)
```

# Q5. Take a number from the user. Calculate its factorial using a while loop.

```python
num = int(input("Enter number: "))
i = 1
fact = 1
while i <= num:
  fact *= i
  i += 1
print('Factorial :',fact)
```

# Q6. Take a number from the user. Reverse the digits of the number using a while loop.

 Example:
 - Input: 1234
- Output: 4321




```python
num = int(input('Enter number: '))
rev = 0
while num > 0:
  digit = num % 10
  rev = rev*10+digit
  num = num//10

print('Reverse :',rev)
```

# Q7. Take a number from the user. Count the total number of digits using a while loop.
 Example:
- Input: 98765
- Output: 5


```python
num = int(input('Enter number: '))
count = 0
while num > 0:
  count += 1
  num = num//10
print('total numbers:',count)
```

# Q8. Take a number from the user. Find the sum of its digits.
 Example:
- Input: 1234
- Output: 10

```python
num = int(input('Enter number: '))
sum_dig = 0
while num > 0:
  sum_dig += num%10
  num //=10

print("Sum of digits:",sum_dig)
```

# Q9. Keep taking numbers from the user until the user enters 0.
 Print:
 - total numbers entered
 - sum of all numbers
 - average of numbers

```python
total = 0
count = 0

num = int(input("Enter number: "))

while num != 0:
    total += num
    count += 1

    num = int(input("Enter number: "))

print("======================================")
print("Total numbers entered:", count)
print("Sum of numbers:", total)

if count > 0:
    print("Average:", total / count)
else:
    print("Average: 0")
```

 Q10. Build a simple ATM menu using a while loop.

 Menu:
 1. Check Balance
 2. Deposit
 3. Withdraw
 4. Exit

 Keep showing the menu until the user selects Exit.

```python
balance = 1000
while True:
  print("\nMenu")
  print("1. Check Balance")
  print("2. Deposit")
  print("3. Withdraw")
  print("4. Exit")

  choice = int(input("Enter your choice: "))
  if choice == 1:
    print('Balance:')
  elif choice == 2:
    amount = float(input("Enter Deposite amount: "))
    balance += amount
    print('Deposite Successful')
    print('Updated Balance:',balance)
  elif choice == 3:
    amount = float(input("Enter Withdraw amount:"))
    if amount <= balance:
      balance -= amount
      print("Withdraw Successful")
      print('Updated Balance:',balance)

    else:
      print("Insufficient balance")

  elif choice == 4:
       print("Thank you for using ATM.")
       break

  else:
      print("Invalid Choice!")
```
