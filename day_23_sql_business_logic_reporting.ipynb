 Q1. Responsibility Identification
     You are given a class that:
     - stores user data
     - validates user input
     - prints user details

     Task:
     - Identify why this violates Single Responsibility Principle
     - Redesign the solution using multiple classes

```python
class UserData:
    user_id = 0

    def __init__(self, name, age, gender):
        UserData.user_id += 1
        self.user_id = UserData.user_id
        self.name = name
        self.age = age
        self.gender = gender

    def get_details(self):
        return {
            'user_id': self.user_id,
            'name': self.name,
            'age': self.age,
            'gender': self.gender
        }
```

```python
r = UserData('ritesh',22,'male')
```

```python
r.get_details()
```

```python
# Single Responsibility Principle
class User:
  def __init__(self,user_id,name,age,gender):
    self.user_id = user_id
    self.name = name
    self.age = age
    self.gender = gender


class UserIdGenerater:
  _user_id = 0

  @classmethod
  def generate_id(cls):
    cls._user_id += 1
    return cls._user_id


class UserFormatter:
  @staticmethod
  def to_dic(user):
    return {
        'user_id':user.user_id,
        'name': user.name,
        'age' : user.age,
        'gender' : user.gender
    }
```

```python
user_id = UserIdGenerater.generate_id()
user = User(user_id,'Ritesh',22,'Male')
```

```python
details = UserFormatter.to_dic(user)
```

```python
details
```


 Q2. Encapsulation Judgment

     Design a class `BankAccount` with:
     - account_number
     - holder_name
     - balance

     Rules:
     - Balance should NOT be modified directly
     - Only deposit() and withdraw() can change balance

     Task:
     - Enforce encapsulation properly


```python
class BankAccount:
  def __init__(self,account_number,holder_name,initial_balance=0):
    self.account_number = account_number
    self.holder_name = holder_name
    self.__balance = initial_balance

  def deposit(self,amount):
    if amount < 0:
      raise ValueError('Deposit amount must be Positive')

    self.__balance += amount
    print('Deposit successful!!')

  def withdraw(self,amount):
    if amount < 0:
      raise ValueError('Withdrawl amount must be Positive')

    if amount > self.__balance:
      raise TypeError('Insufficient amount')

    self.__balance -= amount
    print('Withdrawl successful!!')

  def get_balance(self):
    return self.__balance
```

```python
acc = BankAccount(101, "Ritesh", 5000)
```

```python
acc.deposit(2000)
```

```python
acc.withdraw(1000)
```

```python
acc.get_balance()
```

 Q3. Inheritance vs Composition Decision
     Scenario:

     - You need to model EmailNotification and SMSNotification

     Task:
     - Decide whether inheritance or composition is better
     - Implement your chosen design

```python
class MessageSender:
    def send(self, message):
        raise NotImplementedError
```

```python
class EmailSender(MessageSender):
    def send(self, message):
        print(f"Sending EMAIL: {message}")


class SMSSender(MessageSender):
    def send(self, message):
        print(f"Sending SMS: {message}")
```

```python
class Notification:
    def __init__(self, sender: MessageSender):
        self.sender = sender

    def notify(self, message):
        self.sender.send(message)
```

```python
email_notification = Notification(EmailSender())
email_notification.notify("Welcome to our platform")
```

```python
sms_notification = Notification(SMSSender())
sms_notification.notify("Your OTP is 1234")
```

 Q4. Polymorphism Without Conditionals

     Design a system where different payment methods
     (Card, UPI, NetBanking) process payments.

     Rules:
     - NO if/elif chains based on type
     - Use polymorphism properly

```python
class PaymentMethod:
    def pay(self, amount):
        raise NotImplementedError

class CardPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Processing card payment of ₹{amount}")

class UPIPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Processing UPI payment of ₹{amount}")

class NetBankingPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Processing net banking payment of ₹{amount}")
```

```python
class PaymentProcessor:
    def process_payment(self, payment_method: PaymentMethod, amount):
        payment_method.pay(amount)
```

```python
processor = PaymentProcessor()
```

```python
processor.process_payment(CardPayment(), 1000)
```

```python
processor.process_payment(UPIPayment(), 500)
```

```python
processor.process_payment(NetBankingPayment(), 2000)
```

 Q5. Constructor Design Trap

     Create a class where:
     - Some attributes are mandatory
     - Some attributes are optional

     Task:
     - Design a clean __init__ method
     - Avoid too many parameters
     - Explain how poor constructor design hurts usability

```python
class User:
    def __init__(self, user_id, name, email, phone, address, age, gender):
        self.user_id = user_id
        self.name = name
        self.email = email
        self.phone = phone
        self.address = address
        self.age = age
        self.gender = gender
```

```python
class User:
    def __init__(self, user_id, name, email, *, phone=None, address=None, age=None):
        self.user_id = user_id      # mandatory
        self.name = name            # mandatory
        self.email = email          # mandatory
        self.phone = phone          # optional
        self.address = address      # optional
        self.age = age              # optional
```

```python
u1 = User(1, "Ritesh", "ritesh@email.com")
```

```python
u2 = User(
    2,
    "Aman",
    "aman@email.com",
    phone="9876543210",
    age=25
)
```
