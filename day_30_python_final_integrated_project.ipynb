```python
import json


class Student:

    def __init__(self, student_id, name, age, course, marks):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.course = course
        self.marks = marks

    def total_marks(self):
        return sum(self.marks.values())

    def percentage(self):
        return self.total_marks() / 500 * 100

    def grade(self):

        percentage = self.percentage()

        if percentage >= 90:
            return "A"

        elif percentage >= 75:
            return "B"

        elif percentage >= 60:
            return "C"

        elif percentage >= 40:
            return "D"

        else:
            return "F"

    def __str__(self):
        return (
            f"ID: {self.student_id} | "
            f"Name: {self.name} | "
            f"Course: {self.course} | "
            f"Percentage: {self.percentage():.2f}% | "
            f"Grade: {self.grade()}"
        )


class StudentManager:

    def __init__(self):
        self.students = []

    def add_student(self, student):

        for s in self.students:
            if s.student_id == student.student_id:
                print("Duplicate Student ID")
                return

        self.students.append(student)
        print("Student Added Successfully")

    def display_students(self):

        if len(self.students) == 0:
            print("No Students Found")
            return

        for student in self.students:
            print(student)

    def search_by_id(self, student_id):

        for student in self.students:
            if student.student_id == student_id:
                print(student)
                return

        print("Student Not Found")

    def search_by_name(self, name):

        for student in self.students:
            if student.name.lower() == name.lower():
                print(student)
                return

        print("Student Not Found")

    def update_student(self, student_id):

        for student in self.students:

            if student.student_id == student_id:

                student.name = input("Enter New Name: ")
                student.age = int(input("Enter New Age: "))
                student.course = input("Enter New Course: ")

                print("Student Updated Successfully")
                return

        print("Student Not Found")

    def update_marks(self, student_id):

        for student in self.students:

            if student.student_id == student_id:

                student.marks["Python"] = int(input("Python: "))
                student.marks["SQL"] = int(input("SQL: "))
                student.marks["Statistics"] = int(input("Statistics: "))
                student.marks["Machine Learning"] = int(input("Machine Learning: "))
                student.marks["Power BI"] = int(input("Power BI: "))

                print("Marks Updated Successfully")
                return

        print("Student Not Found")

    def delete_student(self, student_id):

        for student in self.students:

            if student.student_id == student_id:
                self.students.remove(student)

                print("Student Deleted Successfully")
                return

        print("Student Not Found")

    def topper(self):

        if len(self.students) == 0:
            return

        topper = max(
            self.students,
            key=lambda student: student.percentage()
        )

        print("\nTOPPER")
        print(topper)

    def lowest_scorer(self):

        if len(self.students) == 0:
            return

        lowest = min(
            self.students,
            key=lambda student: student.percentage()
        )

        print("\nLOWEST SCORER")
        print(lowest)

    def average_marks(self):

        if len(self.students) == 0:
            return

        avg = sum(
            student.percentage()
            for student in self.students
        ) / len(self.students)

        print("Average Percentage:", round(avg, 2))

    def grade_distribution(self):

        grades = {
            "A": 0,
            "B": 0,
            "C": 0,
            "D": 0,
            "F": 0
        }

        for student in self.students:
            grades[student.grade()] += 1

        print(grades)

    def pass_percentage(self):

        if len(self.students) == 0:
            return

        passed = 0

        for student in self.students:

            if student.grade() != "F":
                passed += 1

        percentage = passed / len(self.students) * 100

        print("Pass Percentage:", round(percentage, 2))

    def save_file(self):

        data = []

        for student in self.students:

            data.append(
                {
                    "student_id": student.student_id,
                    "name": student.name,
                    "age": student.age,
                    "course": student.course,
                    "marks": student.marks
                }
            )

        with open("students.json", "w") as file:
            json.dump(data, file, indent=4)

        print("Data Saved Successfully")

    def load_file(self):

        try:

            with open("students.json", "r") as file:
                data = json.load(file)

            self.students = []

            for item in data:

                student = Student(
                    item["student_id"],
                    item["name"],
                    item["age"],
                    item["course"],
                    item["marks"]
                )

                self.students.append(student)

            print("Data Loaded Successfully")

        except FileNotFoundError:
            print("No File Found")


manager = StudentManager()

while True:

    print("\n===== STUDENT MANAGEMENT SYSTEM =====")

    print("1. Add Student")
    print("2. Display Students")
    print("3. Search By ID")
    print("4. Search By Name")
    print("5. Update Student")
    print("6. Update Marks")
    print("7. Delete Student")
    print("8. Topper")
    print("9. Lowest Scorer")
    print("10. Average Percentage")
    print("11. Grade Distribution")
    print("12. Pass Percentage")
    print("13. Save File")
    print("14. Load File")
    print("15. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        try:

            student_id = int(input("Student ID: "))
            name = input("Name: ")
            age = int(input("Age: "))
            course = input("Course: ")

            marks = {
                "Python": int(input("Python: ")),
                "SQL": int(input("SQL: ")),
                "Statistics": int(input("Statistics: ")),
                "Machine Learning": int(input("Machine Learning: ")),
                "Power BI": int(input("Power BI: "))
            }

            student = Student(
                student_id,
                name,
                age,
                course,
                marks
            )

            manager.add_student(student)

        except ValueError:
            print("Invalid Input")

    elif choice == "2":
        manager.display_students()

    elif choice == "3":
        manager.search_by_id(
            int(input("Enter Student ID: "))
        )

    elif choice == "4":
        manager.search_by_name(
            input("Enter Name: ")
        )

    elif choice == "5":
        manager.update_student(
            int(input("Student ID: "))
        )

    elif choice == "6":
        manager.update_marks(
            int(input("Student ID: "))
        )

    elif choice == "7":
        manager.delete_student(
            int(input("Student ID: "))
        )

    elif choice == "8":
        manager.topper()

    elif choice == "9":
        manager.lowest_scorer()

    elif choice == "10":
        manager.average_marks()

    elif choice == "11":
        manager.grade_distribution()

    elif choice == "12":
        manager.pass_percentage()

    elif choice == "13":
        manager.save_file()

    elif choice == "14":
        manager.load_file()

    elif choice == "15":
        print("Exiting...")
        break

    else:
        print("Invalid Choice")
```
