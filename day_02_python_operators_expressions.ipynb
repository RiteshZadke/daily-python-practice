# Q1. Take two integers from the user and perform:
 - Addition
 - Subtraction
 - Multiplication
 - Division
 - Floor Division
 - Modulus
 - Exponentiation


```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

print(f'Addition: {num1+num2}')
print(f'Subtraction: {num1-num2}')
print(f'Multiplication: {num1*num2}')
print(f'Division: {num1/num2}')
print(f'Floor Division: {num1//num2}')
print(f'Modulus: {num1%num2}')
print(f'Exponentiation: {num1**num2}')
```

# Q3. Take three numbers from the user.
 Print:
 - largest number
 - smallest number

 using comparison operators only.



```python
num1 = int(input('Enter first number: '))
num2 = int(input('Enter second number: '))
num3 = int(input('Enter third number: '))

largest = num1

if num2 > largest:
  largest = num2
if num3 > largest:
  largest = num3

smallest = num1

if smallest > num2:
  smallest = num2
if smallest > num3:
  smallest = num3
print('==============================')
print('largest number: ',largest)
print('smallest number: ',smallest)
```

# Q4. Take marks of 3 subjects.
 Calculate:
 - total marks
 - average marks

 Then check whether average is greater than 75.



```python
os = int(input('Enter marks of OS: '))
py = int(input('Enter marks of py: '))
sql = int(input('Enter marks of sql: '))

total = os+py+sql
average = total/3
print('=========================================')
print('Total marks: ',total)
print('Average marks: ',average)
print('Average greater than 75: ',average > 75)
```


# Q5. Take user age as input.
 Check:
 - age is greater than or equal to 18
 - age is less than 60

 Print final eligibility using logical operators.



```python
age = int(input('Enter your age: '))
is_eligible = (age >= 18) and (age < 60)
print("Age greater than or equal to 18:", age >= 18)
print("Age less than 60:", age < 60)

print("Final Eligibility:", is_eligible)
```

# Q8. Take two boolean values from the user.

 - AND operation

 - NOT operation




```python
bol1 = input('Enter first boolean value (True/False): ') == 'True'
bol2 = input('Enter second boolean value (True/False): ') == 'True'

print('AND operation',bol1 and bol2)
print('OR operation',bol1 or bol2)
print('NOT operation on first value:', not bol1)
print('NOT operation on second value:', not bol2)
```

# Q9. Take a number from the user. Perform shorthand assignment operations:
 - +=

 - *=
 - /=

 Print value after each operation.




```python
num = int(input('Enter any number: '))
num += 10
print('+= :',num)
num -= 10
print('-= :',num)
num *= 10
print('*= :',num)
num /= 10
print('/= :',num)
```
