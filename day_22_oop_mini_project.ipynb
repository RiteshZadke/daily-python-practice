 Create a class BankAccount
 Attributes:
 - account_number
 - account_holder
 - _balance (protected)

 Methods:
 - deposit(amount)
 - withdraw(amount)
 - get_balance()

 Rules:
 - amount must be positive
 - balance must never go negative
 - invalid operations raise ValueError


```python
class BankAccount:
    def __init__(self, acc_number, acc_holder):
        self.acc_number = acc_number
        self.acc_holder = acc_holder
        self._balance = 0

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")

        self._balance += amount
        return "Deposit successful"

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")

        if amount > self._balance:
            raise ValueError("Insufficient balance")

        self._balance -= amount
        return "Withdrawal successful"

    def get_balance(self):
        return self._balance
```

```python
acc = BankAccount(101, "Ritesh")
```

```python
acc.deposit(500)
```

```python
acc.withdraw(200)
```

```python
print(acc.get_balance())
```

 Create SavingsAccount inheriting from BankAccount

 Additional attributes:
 - interest_rate

 Additional method:
 - apply_interest()

 Use super() correctly


```python
class SavingsAccount(BankAccount):
  def __init__(self,acc_number,acc_holder,interest_rate):
    super().__init__(acc_number,acc_holder)

    if interest_rate < 0:
      raise ValueError('Interest rate cannot be negetive')
    self.interest_rate = interest_rate

  def apply_interest(self):
    if self._balance <= 0:
      raise ValueError('Interest cannot be applied to zero or negative balance')

    interest = self._balance * (self.interest_rate/100)
    self._balance += interest
    return self._balance
```

```python
s = SavingsAccount(123,'Ram',4)
```

```python
s.deposit(10000)
```

```python
s.apply_interest()
```
