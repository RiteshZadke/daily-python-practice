# Q1. Create a program that finds the second largest number in a list without using sort().




```python
num = [22,45,34,55,49,12,56]
sec_largest  = largest = float('-inf')
for i in num:
  if i > largest:
    sec_largest = largest
    largest = i

print('Second largest number:',sec_largest)
```

```python
text1 = "".join(sorted(input("Enter first word: ").lower()))
text2 = "".join(sorted(input("Enter second word: ").lower()))

if text1 == text2:
  print("Anagram")

else:
  print('Not anagram')
```

# Q3. Create a program that finds all duplicate elements in a list.
 Display:
 - duplicate values
 - frequency of each duplicate




```python
numbers = [10, 20, 30, 20, 40, 10, 50, 20, 30]
freq = {}
for num in numbers:
  freq[num] = freq.get(num,0)+1

for key,value in freq.items():
  if value > 1:
    print(f'Number: {key}, frequency: {value}')
```

# Q5. Create a program that finds the first non-repeating character in a string.

 Example:
 programming

 Output:
 p


```python
text = input("Enter string: ")
freq = {}
for i in text:
  freq[i] = freq.get(i,0)+1

for char in text:
    if freq[char] == 1:
        print("First non-repeating character:", char)
        break
```

# Q6. Create a program that checks whether a number is an Armstrong number.

 Example:
- 153
- 370
- 371




```python
number = int(input("Enter a number: "))
original = number
digits = len(str(number))
total = 0

while number > 0:
    digit = number % 10
    total += digit ** digits
    number //= 10

if total == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

# Q7. Create a program that merges two sorted lists into a single sorted list without using sort().

```python
list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]

merged = []
i = 0
j = 0

while i < len(list1) and j < len(list2):
    if list1[i] < list2[j]:
        merged.append(list1[i])
        i += 1
    else:
        merged.append(list2[j])
        j += 1

while i < len(list1):
    merged.append(list1[i])
    i += 1

while j < len(list2):
    merged.append(list2[j])
    j += 1

print("Merged List:", merged)
```

# Q8. Create a program that finds the most frequently occurring element in a list.

 Display:
 - element
 - frequency




```python
numbers = [10, 20, 30, 20, 40, 10, 50, 20, 30]
freq = {}
most_occ = float('-inf')
high = ''
for num in numbers:
  freq[num] = freq.get(num,0)+1

for key,value in freq.items():
  if value > most_occ:
    high = key
    most_occ = value

print(f"Element: {high}, frequency: {most_occ}")
```

# Q9. Create a program that groups words by their length.

 Example:
 - ['cat','dog','python','java']

 Output:
 - {
   3:['cat','dog'],
   4:['java'],
   6:['python']
 }


```python
l1 = ['cat','dog','python','java']
freq = {}
for i in l1:
  length = len(i)
  freq[length] = freq.get(length,[])+ [i]

print('Grouped:',freq)
```

# Q10. Build a Student Analytics System.

 Input:
 - student name
 - marks in 5 subjects

 Generate:
 - total
 - percentage
 - grade
 - rank list
 - topper details
 - subject-wise highest marks

```python
students = []

num_students = int(input("Enter number of students: "))

for i in range(num_students):

    name = input(f"\nEnter Student {i+1} Name: ")

    marks = []

    for subject in range(1, 6):

        mark = int(input(f"Enter Subject {subject} Marks: "))
        marks.append(mark)

    total = sum(marks)

    percentage = total / 5

    if percentage >= 90:
        grade = "A"

    elif percentage >= 75:
        grade = "B"

    elif percentage >= 60:
        grade = "C"

    elif percentage >= 40:
        grade = "D"

    else:
        grade = "F"

    students.append({
        "name": name,
        "marks": marks,
        "total": total,
        "percentage": percentage,
        "grade": grade
    })

# Rank List
rank_list = sorted(
    students,
    key=lambda student: student["total"],
    reverse=True
)

# Topper
topper = rank_list[0]

# Subject-wise Highest Marks
subject_highest = []

for subject_index in range(5):

    highest = max(
        student["marks"][subject_index]
        for student in students
    )

    subject_highest.append(highest)

# Display Student Report
print("\n===== STUDENT ANALYTICS REPORT =====")

for student in students:

    print("\nName:", student["name"])
    print("Marks:", student["marks"])
    print("Total:", student["total"])
    print("Percentage:", round(student["percentage"], 2))
    print("Grade:", student["grade"])

# Display Rank List
print("\n===== RANK LIST =====")

for rank, student in enumerate(rank_list, start=1):

    print(
        rank,
        student["name"],
        "- Total:",
        student["total"]
    )

# Display Topper
print("\n===== TOPPER DETAILS =====")

print("Name:", topper["name"])
print("Total:", topper["total"])
print("Percentage:", round(topper["percentage"], 2))
print("Grade:", topper["grade"])

# Display Subject-wise Highest Marks
print("\n===== SUBJECT-WISE HIGHEST MARKS =====")

for subject, marks in enumerate(subject_highest, start=1):

    print(f"Subject {subject}: {marks}")
```
