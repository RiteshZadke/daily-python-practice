# Q4. Create a program that checks whether a number is:

 - prime
 - palindrome
 - Armstrong

 Display all applicable results.

```python
number = int(input("Enter a number: "))

results = []

# Prime Check
if number > 1:

    is_prime = True

    for i in range(2, int(number ** 0.5) + 1):

        if number % i == 0:

            is_prime = False

            break

    if is_prime:
        results.append("Prime")

    else:
      results.append("Not Prime")

if str(number) == str(number)[::-1]:
    results.append("Palindrome")
if  str(number) != str(number)[::-1]:
  results.append("Not Palindrome")

digits = len(str(number))

armstrong_sum = 0

temp = number

while temp > 0:

    digit = temp % 10

    armstrong_sum += digit ** digits

    temp //= 10

if armstrong_sum == number:
    results.append("Armstrong")


if results:

    for result in results:
        print(result)

else:

    print("None")
```

# Q5. Create a program that counts frequency of every character in a string using a dictionary.

```python
text = input("Enter String: ")

frequency = {}

for char in text:

    frequency[char] = (
        frequency.get(char, 0) + 1
    )

print("\nCharacter Frequency")

for key, value in frequency.items():

    print(
        f"{key} : {value}"
    )
```

# Q7. Create a menu-driven calculator.

 Operations:
 - Addition
 - Subtraction
 - Multiplication
 - Division

 Use functions and exception handling.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

while True:
    print("\n===== CALCULATOR =====")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = input("Enter Choice: ")

    if choice == "5":
        break

    try:
        num1 = float(input("Enter First Number: "))
        num2 = float(input("Enter Second Number: "))

        if choice == "1":
            print("Result:", add(num1, num2))

        elif choice == "2":
            print("Result:", subtract(num1, num2))

        elif choice == "3":
            print("Result:", multiply(num1, num2))

        elif choice == "4":
            print("Result:", divide(num1, num2))

        else:
            print("Invalid Choice")

    except ValueError:
        print("Error: Enter Numeric Values")

    except ZeroDivisionError:
        print("Error: Division By Zero")
```

# Q8. Create a program that accepts employee data.

 Store:
 - employee_id
 - employee_name
 - salary

 Generate:
 - highest salary
 - lowest salary
 - average salary

```python
employees = []

n = int(input("Enter Number Of Employees: "))

for i in range(n):
    employee_id = input("Enter Employee ID: ")
    employee_name = input("Enter Employee Name: ")
    salary = float(input("Enter Salary: "))

    employees.append({
        "id": employee_id,
        "name": employee_name,
        "salary": salary
    })

highest_salary = max(employees, key=lambda employee: employee["salary"])
lowest_salary = min(employees, key=lambda employee: employee["salary"])

average_salary = (
    sum(employee["salary"] for employee in employees)
    / len(employees)
)

print("\n===== REPORT =====")
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)
print("Average Salary:", average_salary)
```

# Q9. Create a Contact Book Application.

 Features:
 - Add Contact
 - Search Contact
 - Update Contact
 - Delete Contact
 - View All Contacts

 Use dictionaries.

```python
contacts = []

def add_contact():
    name = input("Enter Name: ")
    phone = input("Enter Phone: ")
    email = input("Enter Email: ")

    contacts.append({
        "name": name,
        "phone": phone,
        "email": email
    })

    print("Contact Added")

def search_contact():
    target = input("Enter Name: ")

    for contact in contacts:
        if contact["name"].lower() == target.lower():
            print(contact)
            return

    print("Contact Not Found")

def update_contact():
    target = input("Enter Name: ")

    for contact in contacts:
        if contact["name"].lower() == target.lower():
            contact["phone"] = input("Enter New Phone: ")
            contact["email"] = input("Enter New Email: ")

            print("Contact Updated")
            return

    print("Contact Not Found")

def delete_contact():
    target = input("Enter Name: ")

    for contact in contacts:
        if contact["name"].lower() == target.lower():
            contacts.remove(contact)

            print("Contact Deleted")
            return

    print("Contact Not Found")

def view_contacts():
    if not contacts:
        print("No Contacts Found")
        return

    for contact in contacts:
        print(contact)

while True:
    print("\n===== CONTACT BOOK =====")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. View Contacts")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_contact()

    elif choice == "2":
        search_contact()

    elif choice == "3":
        update_contact()

    elif choice == "4":
        delete_contact()

    elif choice == "5":
        view_contacts()

    elif choice == "6":
        break

    else:
        print("Invalid Choice")
```

# Q10. Build a Mini Management System.

 Combine:
 - Functions
 - Lists
 - Dictionaries
 - File Handling
 - Exception Handling

 Features:
 - Add Record
 - View Records
 - Search Record
 - Update Record
 - Delete Record

 Generate summary statistics.

```python
FILE_NAME = "records.txt"

def add_record():
    record_id = input("Enter Record ID: ")
    name = input("Enter Name: ")
    value = float(input("Enter Value: "))

    with open(FILE_NAME, "a") as file:
        file.write(f"{record_id},{name},{value}\n")

    print("Record Added Successfully")

def view_records():
    try:
        with open(FILE_NAME, "r") as file:
            for line in file:
                print(line.strip())

    except FileNotFoundError:
        print("No Records Found")

def search_record():
    target_id = input("Enter Record ID: ")

    try:
        with open(FILE_NAME, "r") as file:
            for line in file:
                record_id, name, value = line.strip().split(",")

                if record_id == target_id:
                    print(line.strip())
                    return

        print("Record Not Found")

    except FileNotFoundError:
        print("No Records Found")

def update_record():
    target_id = input("Enter Record ID: ")

    records = []
    found = False

    try:
        with open(FILE_NAME, "r") as file:
            for line in file:
                record_id, name, value = line.strip().split(",")

                if record_id == target_id:
                    name = input("Enter New Name: ")
                    value = input("Enter New Value: ")
                    found = True

                records.append(f"{record_id},{name},{value}\n")

        with open(FILE_NAME, "w") as file:
            file.writelines(records)

        if found:
            print("Record Updated")
        else:
            print("Record Not Found")

    except FileNotFoundError:
        print("No Records Found")

def delete_record():
    target_id = input("Enter Record ID: ")

    records = []
    found = False

    try:
        with open(FILE_NAME, "r") as file:
            for line in file:
                record_id, name, value = line.strip().split(",")

                if record_id == target_id:
                    found = True
                else:
                    records.append(line)

        with open(FILE_NAME, "w") as file:
            file.writelines(records)

        if found:
            print("Record Deleted")
        else:
            print("Record Not Found")

    except FileNotFoundError:
        print("No Records Found")

def summary_statistics():
    values = []

    try:
        with open(FILE_NAME, "r") as file:
            for line in file:
                record_id, name, value = line.strip().split(",")
                values.append(float(value))

        if values:
            print("\n===== SUMMARY =====")
            print("Total Records:", len(values))
            print("Maximum Value:", max(values))
            print("Minimum Value:", min(values))
            print("Average Value:", sum(values) / len(values))

    except FileNotFoundError:
        print("No Records Found")

while True:
    print("\n===== MINI MANAGEMENT SYSTEM =====")
    print("1. Add Record")
    print("2. View Records")
    print("3. Search Record")
    print("4. Update Record")
    print("5. Delete Record")
    print("6. Summary Statistics")
    print("7. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_record()

    elif choice == "2":
        view_records()

    elif choice == "3":
        search_record()

    elif choice == "4":
        update_record()

    elif choice == "5":
        delete_record()

    elif choice == "6":
        summary_statistics()

    elif choice == "7":
        break

    else:
        print("Invalid Choice")
```
