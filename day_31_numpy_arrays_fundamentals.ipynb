# Q1. Import NumPy and create arrays using:

- Python List
- Python Tuple

Display:
- array
- shape
- size
- ndim
- dtype

```python
import numpy as np

# Array from Python List
list_array = np.array([10, 20, 30, 40, 50])

print("Array from List:")
print(list_array)
print("Shape:", list_array.shape)
print("Size:", list_array.size)
print("Dimensions:", list_array.ndim)
print("Data Type:", list_array.dtype)

print("-" * 40)

# Array from Python Tuple
tuple_array = np.array((1, 2, 3, 4, 5))

print("Array from Tuple:")
print(tuple_array)
print("Shape:", tuple_array.shape)
print("Size:", tuple_array.size)
print("Dimensions:", tuple_array.ndim)
print("Data Type:", tuple_array.dtype)
```
