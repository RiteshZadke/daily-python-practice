# Project 1. Student Management System

 Features:
 - Add student
 - View all students
 - Search student by ID
 - Update student marks
 - Delete student record

 Store:
 - student_id
 - student_name
 - marks

 Display:
 - topper
 - average marks
 - grade report




```python
FILE_NAME = "students.txt"


def add_student():

    student_id = input("Enter Student ID: ")
    name = input("Enter Student Name: ")
    marks = input("Enter Student Marks: ")

    with open(FILE_NAME, "a") as file:
        file.write(f"{student_id},{name},{marks}\n")

    print("Student Added Successfully")


def view_students():

    try:
        with open(FILE_NAME, "r") as file:

            records = file.readlines()

            if not records:
                print("No Records Found")
                return

            print("\nStudent Records")

            for record in records:
                student_id, name, marks = record.strip().split(",")

                print(
                    "ID:", student_id,
                    "| Name:", name,
                    "| Marks:", marks
                )

    except FileNotFoundError:
        print("No Student File Found")


def search_student():

    target_id = input("Enter Student ID to Search: ")

    found = False

    try:
        with open(FILE_NAME, "r") as file:

            for record in file:

                student_id, name, marks = record.strip().split(",")

                if student_id == target_id:

                    print("\nStudent Found")
                    print("ID:", student_id)
                    print("Name:", name)
                    print("Marks:", marks)

                    found = True
                    break

        if not found:
            print("Student Not Found")

    except FileNotFoundError:
        print("No Student File Found")


def update_marks():

    target_id = input("Enter Student ID: ")

    updated_records = []

    found = False

    try:
        with open(FILE_NAME, "r") as file:

            for record in file:

                student_id, name, marks = record.strip().split(",")

                if student_id == target_id:

                    new_marks = input("Enter New Marks: ")

                    updated_records.append(
                        f"{student_id},{name},{new_marks}\n"
                    )

                    found = True

                else:
                    updated_records.append(record)

        with open(FILE_NAME, "w") as file:
            file.writelines(updated_records)

        if found:
            print("Marks Updated Successfully")
        else:
            print("Student Not Found")

    except FileNotFoundError:
        print("No Student File Found")


def delete_student():

    target_id = input("Enter Student ID: ")

    updated_records = []

    found = False

    try:
        with open(FILE_NAME, "r") as file:

            for record in file:

                student_id, name, marks = record.strip().split(",")

                if student_id == target_id:
                    found = True

                else:
                    updated_records.append(record)

        with open(FILE_NAME, "w") as file:
            file.writelines(updated_records)

        if found:
            print("Student Deleted Successfully")
        else:
            print("Student Not Found")

    except FileNotFoundError:
        print("No Student File Found")


while True:

    print("\n===== STUDENT FILE MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Student Marks")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        view_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        update_marks()

    elif choice == "5":
        delete_student()

    elif choice == "6":
        print("Program Closed")
        break

    else:
        print("Invalid Choice")
```

# Project 2. Bank Account Management System

 Features:
 - Create account
 - Deposit money
 - Withdraw money
 - Check balance
 - Transaction history

 Validate:
 - insufficient balance
 - invalid amount



```python
accounts = {}


def create_account():

    account_number = input("Enter Account Number: ")
    account_holder = input("Enter Account Holder Name: ")

    accounts[account_number] = {
        "name": account_holder,
        "balance": 0,
        "history": []
    }

    print("Account Created Successfully")


def deposit_money():

    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account Not Found")
        return

    amount = float(input("Enter Deposit Amount: "))

    if amount <= 0:
        print("Invalid Amount")
        return

    accounts[account_number]["balance"] += amount

    accounts[account_number]["history"].append(
        f"Deposited ₹{amount}"
    )

    print("Amount Deposited Successfully")


def withdraw_money():

    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account Not Found")
        return

    amount = float(input("Enter Withdrawal Amount: "))

    if amount <= 0:
        print("Invalid Amount")
        return

    if amount > accounts[account_number]["balance"]:
        print("Insufficient Balance")
        return

    accounts[account_number]["balance"] -= amount

    accounts[account_number]["history"].append(
        f"Withdrawn ₹{amount}"
    )

    print("Amount Withdrawn Successfully")


def check_balance():

    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account Not Found")
        return

    print(
        f"Current Balance: ₹{accounts[account_number]['balance']}"
    )


def transaction_history():

    account_number = input("Enter Account Number: ")

    if account_number not in accounts:
        print("Account Not Found")
        return

    print("\n===== TRANSACTION HISTORY =====")

    history = accounts[account_number]["history"]

    if not history:
        print("No Transactions Found")
        return

    for transaction in history:
        print(transaction)


while True:

    print("\n===== BANK ACCOUNT MANAGEMENT SYSTEM =====")

    print("1. Create Account")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Check Balance")
    print("5. Transaction History")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        create_account()

    elif choice == "2":

        deposit_money()

    elif choice == "3":

        withdraw_money()

    elif choice == "4":

        check_balance()

    elif choice == "5":

        transaction_history()

    elif choice == "6":

        print("Program Closed")
        break

    else:

        print("Invalid Choice")
```

# Project 5. Library Management System

 Features:
 - Add book
 - Issue book
 - Return book
 - Search book

 Store:
 - book_id
 - title
 - author
 - availability

 Generate library report.




```python
books = []


def add_book():

    book_id = input("Enter Book ID: ")
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)

    print("Book Added Successfully")


def issue_book():

    book_id = input("Enter Book ID: ")

    for book in books:

        if book["id"] == book_id:

            if book["available"]:

                book["available"] = False

                print("Book Issued Successfully")

            else:

                print("Book Already Issued")

            return

    print("Book Not Found")


def return_book():

    book_id = input("Enter Book ID: ")

    for book in books:

        if book["id"] == book_id:

            if not book["available"]:

                book["available"] = True

                print("Book Returned Successfully")

            else:

                print("Book Already Available")

            return

    print("Book Not Found")


def search_book():

    book_id = input("Enter Book ID: ")

    for book in books:

        if book["id"] == book_id:

            print("\nBook Found")

            print(f"ID: {book['id']}")
            print(f"Title: {book['title']}")
            print(f"Author: {book['author']}")
            print(
                f"Status: {'Available' if book['available'] else 'Issued'}"
            )

            return

    print("Book Not Found")


def library_report():

    if not books:

        print("No Books Available")

        return

    total_books = len(books)

    available_books = 0

    issued_books = 0

    print("\n===== LIBRARY REPORT =====")

    for book in books:

        print(
            f"ID: {book['id']} | "
            f"Title: {book['title']} | "
            f"Author: {book['author']} | "
            f"Status: {'Available' if book['available'] else 'Issued'}"
        )

        if book["available"]:
            available_books += 1
        else:
            issued_books += 1

    print("\nSummary")
    print("Total Books:", total_books)
    print("Available Books:", available_books)
    print("Issued Books:", issued_books)


while True:

    print("\n===== LIBRARY MANAGEMENT SYSTEM =====")

    print("1. Add Book")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Search Book")
    print("5. Library Report")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        add_book()

    elif choice == "2":

        issue_book()

    elif choice == "3":

        return_book()

    elif choice == "4":

        search_book()

    elif choice == "5":

        library_report()

    elif choice == "6":

        print("Program Closed")

        break

    else:

        print("Invalid Choice")
```

# Project 6. Expense Tracker

 Features:
 - Add expense
 - View expenses
 - Category-wise summary

 Categories:
 - Food
 - Travel
 - Shopping
 - Bills

 Generate monthly expense report.


```python
expenses = []


def add_expense():

    category = input(
        "Enter Category (Food/Travel/Shopping/Bills): "
    )

    amount = float(input("Enter Amount: "))

    expense = {
        "category": category,
        "amount": amount
    }

    expenses.append(expense)

    print("Expense Added Successfully")


def view_expenses():

    if not expenses:

        print("No Expenses Found")

        return

    print("\n===== EXPENSES =====")

    for expense in expenses:

        print(
            f"Category: {expense['category']} | "
            f"Amount: ₹{expense['amount']}"
        )


def category_summary():

    if not expenses:

        print("No Expenses Found")

        return

    summary = {}

    for expense in expenses:

        category = expense["category"]

        summary[category] = (
            summary.get(category, 0)
            + expense["amount"]
        )

    print("\n===== CATEGORY-WISE SUMMARY =====")

    for category, amount in summary.items():

        print(f"{category}: ₹{amount}")


def monthly_report():

    if not expenses:

        print("No Expenses Found")

        return

    total_expense = 0

    summary = {}

    for expense in expenses:

        total_expense += expense["amount"]

        category = expense["category"]

        summary[category] = (
            summary.get(category, 0)
            + expense["amount"]
        )

    highest_category = max(
        summary,
        key=summary.get
    )

    print("\n===== MONTHLY EXPENSE REPORT =====")

    print(f"Total Expenses: ₹{total_expense}")

    print(
        f"Highest Spending Category: "
        f"{highest_category}"
    )

    print(
        f"Amount Spent: "
        f"₹{summary[highest_category]}"
    )

    print("\nCategory Breakdown")

    for category, amount in summary.items():

        print(f"{category}: ₹{amount}")


while True:

    print("\n===== EXPENSE TRACKER =====")

    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Category-wise Summary")
    print("4. Monthly Expense Report")
    print("5. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        add_expense()

    elif choice == "2":

        view_expenses()

    elif choice == "3":

        category_summary()

    elif choice == "4":

        monthly_report()

    elif choice == "5":

        print("Program Closed")

        break

    else:

        print("Invalid Choice")
```

# Project 7. Contact Book Application

 Features:
 - Add contact
 - Search contact
 - Update contact
 - Delete contact

 Store:
 - name
 - phone
 - email




```python
contacts = []


def add_contact():

    name = input("Enter Name: ")
    phone = input("Enter Phone Number: ")
    email = input("Enter Email: ")

    contact = {
        "name": name,
        "phone": phone,
        "email": email
    }

    contacts.append(contact)

    print("Contact Added Successfully")


def search_contact():

    name = input("Enter Name to Search: ")

    for contact in contacts:

        if contact["name"].lower() == name.lower():

            print("\nContact Found")

            print(f"Name: {contact['name']}")
            print(f"Phone: {contact['phone']}")
            print(f"Email: {contact['email']}")

            return

    print("Contact Not Found")


def update_contact():

    name = input("Enter Name to Update: ")

    for contact in contacts:

        if contact["name"].lower() == name.lower():

            contact["phone"] = input("Enter New Phone Number: ")
            contact["email"] = input("Enter New Email: ")

            print("Contact Updated Successfully")

            return

    print("Contact Not Found")


def delete_contact():

    name = input("Enter Name to Delete: ")

    for contact in contacts:

        if contact["name"].lower() == name.lower():

            contacts.remove(contact)

            print("Contact Deleted Successfully")

            return

    print("Contact Not Found")


def view_contacts():

    if not contacts:

        print("No Contacts Found")

        return

    print("\n===== CONTACT LIST =====")

    for contact in contacts:

        print(
            f"Name: {contact['name']} | "
            f"Phone: {contact['phone']} | "
            f"Email: {contact['email']}"
        )


while True:

    print("\n===== CONTACT BOOK APPLICATION =====")

    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. View All Contacts")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        add_contact()

    elif choice == "2":

        search_contact()

    elif choice == "3":

        update_contact()

    elif choice == "4":

        delete_contact()

    elif choice == "5":

        view_contacts()

    elif choice == "6":

        print("Program Closed")

        break

    else:

        print("Invalid Choice")
```

# Project 8. Quiz Management System

 Features:
 - Multiple questions
 - Score calculation
 - Result generation

 Generate:
 - total score
 - percentage
 - pass/fail status




```python
questions = [
    {
        "question": "What is the capital of India?",
        "answer": "Delhi"
    },
    {
        "question": "Which language is used for Data Science?",
        "answer": "Python"
    },
    {
        "question": "What is 5 + 5?",
        "answer": "10"
    },
    {
        "question": "Who is known as the Father of Computers?",
        "answer": "Charles Babbage"
    },
    {
        "question": "Which keyword is used to define a function in Python?",
        "answer": "def"
    }
]


score = 0

print("\n===== QUIZ MANAGEMENT SYSTEM =====")

for index, question in enumerate(questions, start=1):

    print(f"\nQuestion {index}")
    print(question["question"])

    user_answer = input("Your Answer: ")

    if user_answer.lower() == question["answer"].lower():

        print("Correct Answer")

        score += 1

    else:

        print(
            f"Wrong Answer | Correct Answer: "
            f"{question['answer']}"
        )

total_questions = len(questions)

percentage = (score / total_questions) * 100

if percentage >= 50:

    result = "PASS"

else:

    result = "FAIL"

print("\n===== RESULT =====")

print("Total Questions:", total_questions)

print("Correct Answers:", score)

print("Wrong Answers:", total_questions - score)

print(f"Percentage: {percentage:.2f}%")

print("Status:", result)
```

# Project 10. E-Commerce Order Management System

 Features:
 - Add customer
 - Place order
 - View orders
 - Calculate total revenue

 Generate:
 - customer summary
 - sales summary
 - top customer report

```python
customers = []
orders = []


def add_customer():

    customer_id = input("Enter Customer ID: ")
    customer_name = input("Enter Customer Name: ")

    customer = {
        "id": customer_id,
        "name": customer_name
    }

    customers.append(customer)

    print("Customer Added Successfully")


def place_order():

    customer_id = input("Enter Customer ID: ")

    customer_exists = False

    for customer in customers:

        if customer["id"] == customer_id:

            customer_exists = True
            break

    if not customer_exists:

        print("Customer Not Found")
        return

    order_id = input("Enter Order ID: ")
    order_amount = float(input("Enter Order Amount: "))

    order = {
        "order_id": order_id,
        "customer_id": customer_id,
        "amount": order_amount
    }

    orders.append(order)

    print("Order Placed Successfully")


def view_orders():

    if not orders:

        print("No Orders Found")
        return

    print("\n===== ORDERS =====")

    for order in orders:

        print(
            f"Order ID: {order['order_id']} | "
            f"Customer ID: {order['customer_id']} | "
            f"Amount: ₹{order['amount']}"
        )


def calculate_total_revenue():

    total_revenue = sum(
        order["amount"]
        for order in orders
    )

    print(f"Total Revenue: ₹{total_revenue}")


def customer_summary():

    if not customers:

        print("No Customers Found")
        return

    print("\n===== CUSTOMER SUMMARY =====")

    for customer in customers:

        total_orders = 0
        total_spent = 0

        for order in orders:

            if order["customer_id"] == customer["id"]:

                total_orders += 1
                total_spent += order["amount"]

        print(
            f"ID: {customer['id']} | "
            f"Name: {customer['name']} | "
            f"Orders: {total_orders} | "
            f"Spent: ₹{total_spent}"
        )


def sales_summary():

    if not orders:

        print("No Orders Found")
        return

    total_orders = len(orders)

    total_revenue = sum(
        order["amount"]
        for order in orders
    )

    average_order_value = (
        total_revenue / total_orders
    )

    print("\n===== SALES SUMMARY =====")

    print("Total Orders:", total_orders)
    print(f"Total Revenue: ₹{total_revenue}")
    print(
        f"Average Order Value: "
        f"₹{average_order_value:.2f}"
    )


def top_customer_report():

    if not customers or not orders:

        print("Insufficient Data")
        return

    customer_spending = {}

    for customer in customers:

        customer_spending[customer["id"]] = 0

    for order in orders:

        customer_spending[
            order["customer_id"]
        ] += order["amount"]

    top_customer_id = max(
        customer_spending,
        key=customer_spending.get
    )

    for customer in customers:

        if customer["id"] == top_customer_id:

            print("\n===== TOP CUSTOMER =====")

            print(
                f"ID: {customer['id']}"
            )

            print(
                f"Name: {customer['name']}"
            )

            print(
                f"Total Spending: "
                f"₹{customer_spending[top_customer_id]}"
            )

            return


while True:

    print(
        "\n===== E-COMMERCE ORDER MANAGEMENT SYSTEM ====="
    )

    print("1. Add Customer")
    print("2. Place Order")
    print("3. View Orders")
    print("4. Calculate Total Revenue")
    print("5. Customer Summary")
    print("6. Sales Summary")
    print("7. Top Customer Report")
    print("8. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        add_customer()

    elif choice == "2":

        place_order()

    elif choice == "3":

        view_orders()

    elif choice == "4":

        calculate_total_revenue()

    elif choice == "5":

        customer_summary()

    elif choice == "6":

        sales_summary()

    elif choice == "7":

        top_customer_report()

    elif choice == "8":

        print("Program Closed")
        break

    else:

        print("Invalid Choice")
```
