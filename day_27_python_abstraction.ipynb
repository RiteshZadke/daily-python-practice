# Q1. Create an abstract class Shape.

 Abstract Method:
 - area()

 Create child classes:
 - Circle
 - Rectangle

 Implement area() in both classes.


```python
from abc import ABC,abstractmethod

class Shape(ABC):
  @abstractmethod
  def area(self):
    pass

class Circle(Shape):
  def __init__(self,radius):
    self.radius = radius

  def area(self):
    return 3.14*self.radius*self.radius

class Rectangle(Shape):
  def __init__(self,length,breath):
    self.lenght = length
    self.breath = breath

  def area(self):
    return self.lenght * self.breath
```

```python
c1 = Circle(7)
r1 = Rectangle(10, 5)
```

```python
print("Circle Area:", c1.area())
print("Rectangle Area:", r1.area())
```

 # Q6. Create an abstract class BankAccount.

 Abstract Methods:
 - deposit()
 - withdraw()

 Create child classes:
 - SavingsAccount
 - CurrentAccount

 Implement account operations.

```python
class BankAccount(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass


class SavingsAccount(BankAccount):
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")


class CurrentAccount(BankAccount):
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        self.balance -= amount
        print("Withdrawn:", amount)
```

```python
s1 = SavingsAccount(10000)
c1 = CurrentAccount(15000)
```

```python
s1.deposit(2000)
s1.withdraw(3000)
```

```python
c1.deposit(5000)
c1.withdraw(2000)
```

```python
from abc import ABC, abstractmethod


class Database(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def disconnect(self):
        pass


class MySQLDatabase(Database):
    def connect(self):
        print("Connected to MySQL")

    def disconnect(self):
        print("Disconnected from MySQL")


class MongoDBDatabase(Database):
    def connect(self):
        print("Connected to MongoDB")

    def disconnect(self):
        print("Disconnected from MongoDB")
```

 # Q10. Build a Banking System.

 Abstract Class:
 - Account

 Abstract Methods:
 - deposit()
 - withdraw()
 - display_account()

 Child Classes:
 - SavingsAccount
 - CurrentAccount

 Create multiple account objects and
 demonstrate abstraction through a
 common interface.

```python
from abc import ABC, abstractmethod


class Account(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

    @abstractmethod
    def display_account(self):
        pass


class SavingsAccount(Account):
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount

    def display_account(self):
        print("Savings Account")
        print("Account No:", self.account_no)
        print("Balance:", self.balance)


class CurrentAccount(Account):
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

    def display_account(self):
        print("Current Account")
        print("Account No:", self.account_no)
        print("Balance:", self.balance)
```

```python
accounts = [
    SavingsAccount(101, 10000),
    CurrentAccount(201, 15000)
]
```

```python
for account in accounts:
    account.deposit(1000)
    account.withdraw(500)
    account.display_account()
    print("-" * 30)
```
